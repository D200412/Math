## 10.2 一元函数：一个自变量

有一个自变量的函数叫做<font color="blue">一元函数</font>(univariate function)。

数据科学和机器学习中常用的函数一般分为<font color="blue">代数函数</font>(algebraic function)和<font color="blue">超越函数</font>(transcendental function)。
- 代数函数是指通过常数与自变量相互之间有限次的加、减、乘、除、有理指数幂和开方等运算构造的函数。绝对值函数也归类到代数函数中。
- 超越函数指的是“超出”代数函数范畴的函数，比如对数函数、指数函数、三角函数等等。

## 10.3 一元函数性质

### 奇偶性

$f(x)$ 若为<font color="blue">偶函数</font>(even function)，对于定义域内任意 $x$ 如下关系都成立：$f(x)=f(-x)$。从几何角度，$f(x)$ 若为偶函数，函数图像关于纵轴对称。

$f(x)$ 若为<font color="blue">奇函数</font>(oddfunction)，对于定义域内任意 $x$ 如下关系都成立： $f(-x)=-f(x)$。从几何角度，$f(x)$ 若为奇函数，函数图像关于原点对称。

### 连续性


简单来说，<font color="blue">连续函数</font>(continuous function)是指函数 $y=f(x)$ 当自变量 $x$ 的变化很小时，所引起的因变量 $y$ 的变化也很小，没有函数值突变；与之相对的就是<font color="blue">不连续函数</font>(discontinuous function)。

不连续函数有几种：<font color="blue">渐近线间断</font>(asymptotic discontinuity)，<font color="blue">点间断</font>(pointdiscontinuity)，还有<font color="blue">跳跃间断</font>(Gump discontinuity)。



### 单调性

```python
from sympy import is_increasing
from sympy.abc import x, y
from sympy import Interval, oo

expr_1 = - x ** 2
print(is_increasing(expr_1, Interval(-oo, 0)))

print(is_increasing(expr_1, Interval(0, +oo)))

expr_2 = -x**2 + y
print(is_increasing(expr_2, Interval(-1, 2), x))
```

### 周期性

函数<font color="blue">周期性</font>(periodicity)。如果函数 $f(x)$ 中不同位置 $x$ 满足 $f(x+T)=f(x)$，则函数为周期函数。$T$ 为<font color="blue">周期</font>(period)。

### 凸凹性

$f(x)$ 在该区间内为凸函数。



$f(x)$ 在该区间内为凹函数。





切线解析式：
$$
y=f(c)+k(x-c)
$$
如果函数为凸函数，当 $x≠c$，函数 $f(x)$ 图像在切线上方：
$$
f(x)>f(c)+k(x-c),\; x \in (a,b),x≠c
$$
如果函数为凹函数，当 $x≠c$，函数 $f(x)$ 图像在切线下方：
$$
f(x)<f(c)+k(x-c),x \in (a,b),x≠c
$$



### 反函数

### 隐函数

<font color="blue">隐函数</font>(implicit function)是由<font color="blue">隐式方程</font>(implicit equation)所隐含定义的函数。

### 变化率和面积

函数变化率叫做<font color="blue">导数</font>(derivative)。

某个函数图形在一定取值范围和横轴围成几何图形的面积，叫做<font color="blue">积分</font>(integral)

## 10.4 二元函数：两个自变量

有两个自变量函数叫做<font color="blue">二元函数</font>(bivariate function)，比如 $y=f(x_1,x_2)$。

此外，有多个自变量函数叫做<font color="blue">多元函数</font>(multivariate function)，比如 $y=(x_1,x_2,...x_D)$。

### 网格化数据

### 一个复杂曲面

复杂函数解析式：
$$
f(x_1,x_2)=3(1-x_1)^2exp(-x_1^2-(x_2+1)^2)-10(\frac{x_1}{5}-x^3_1-x_2^5)exp(-x_1^2-x^2_2)-\frac{1}{3}exp(-(x_1+1)^2-x_2^2)
$$

```python
import numpy as np
from sympy import lambdify, diff, exp, latex, simplify
from sympy.abc import x, y
import matplotlib.pyplot as plt
from matplotlib import cm

num = 301
x_array = np.linspace(-3, 3, num)
y_array = np.linspace(-3, 3, num)
xx, yy = np.meshgrid(x_array, y_array)

plt.close("all")

f_xy = 3 * (1 - x) ** 2 * exp(-(x ** 2) - (y + 1) ** 2) - 10 * (x / 5 - x ** 3 - y ** 5) * exp( -x ** 2 - y ** 2) - (1 / 3) * exp(-(x + 1) ** 2 - y ** 2)

f_xy_fcn = lambdify([x, y], f_xy)
f_xy_zz = f_xy_fcn(xx, yy)

fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz,
                 color=[0.5, 0.5, 0.5],
                 rstride=5, cstride=5,
                 linewidth=0.25)

ax.set_proj_type("ortho")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$f(x,y)$")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.view_init(azim=-135, elev=30)
plt.tight_layout()
ax.grid(False)
plt.show()
```

## 10.5 降维：二元函数切一刀得到一元函数

### $x_1y$ 平面方向剖面线

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz, color=[0.5, 0.5, 0.5], rstride=5, cstride=0, linewidth=0.5)
colors = plt.cm.rainbow(np.linspace(0, 1, 7))

i = 0
for b in [-3, -2, -1, 0, 1, 2, 3]:
    f_xy_b = f_xy.subs(y, b)
    f_xy_b_fcn = lambdify([x], f_xy_b)
    f_xy_b_zz = f_xy_b_fcn(x_array)

    ax.plot(x_array, x_array*0+b, f_xy_b_zz, color=colors[i,:], label= "$x_2$ = %0.0f" %b)
    i += 1

plt.legend()
ax.set_proj_type("ortho")
ax.set_xlabel("$x$")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
plt.tight_layout()
ax.view_init(azim=-135, elev=30)
ax.grid(False)
plt.show()
```

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz, color=[0.5, 0.5, 0.5], linewidth=0.25, rstride=5, cstride=0)
ax.contour(xx, yy, f_xy_zz, levels=60, zdir="y", offset=yy.max(), cmap="rainbow")

ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$f(x,y)$")
ax.set_proj_type("ortho")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.view_init(azim=-135, elev=30)
plt.tight_layout()
ax.grid(False)
plt.show()
```

```python
down_step = 2
y_array_downsample = y_array[0::down_step]

fig, ax = plt.subplots()

colors = plt.cm.rainbow(np.linspace(0, 1, len(y_array_downsample)))

for i in np.linspace(1, len(y_array_downsample), len(y_array_downsample)):
    plt.plot(x_array, f_xy_zz[(int(i) - 1) * down_step, :], color=colors[int(i) - 1])

plt.axhline(y=0, color="k", linestyle="-")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.xlabel("$x_1$")
plt.ylabel("%f(x_1, x_2=b)%")
ax.set_xlim(xx.min(), xx.max())
plt.show()
```

### $x_2y$ 平面方向剖面线

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz, color=[0.5, 0.5, 0.5], rstride=5, cstride=0, linewidth=0.5)
colors = plt.cm.rainbow(np.linspace(0, 1, 7))

i = 0
for a in [-3, -2, -1, 0, 1, 2, 3]:
    f_xy_a = f_xy.subs(x, a)
    f_xy_a_fcn = lambdify([y], f_xy_a)
    f_xy_a_zz = f_xy_a_fcn(y_array)

    ax.plot(y_array*0+a, y_array, f_xy_a_zz, color=colors[i,:], label= "$x_1$ = %0.0f" %a)
    i += 1

plt.legend()
ax.set_proj_type("ortho")
ax.set_xlabel("$x$")
ax.set_ylabel("y")
ax.set_zlabel("f(x,y)")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
plt.tight_layout()
ax.view_init(azim=-135, elev=30)
ax.grid(False)
plt.show()
```

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz, color=[0.5, 0.5, 0.5], linewidth=0.25, rstride=5, cstride=0)
ax.contour(xx, yy, f_xy_zz, levels=60, zdir="x", offset=xx.max(), cmap="rainbow")

ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$f(x,y)$")
ax.set_proj_type("ortho")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.view_init(azim=-135, elev=30)
plt.tight_layout()
ax.grid(False)
plt.show()
```

```python
down_step = 2
x_array_downsample = x_array[0::down_step]

fig, ax = plt.subplots()

colors = plt.cm.rainbow(np.linspace(0, 1, len(x_array_downsample)))

for i in np.linspace(1, len(x_array_downsample), len(x_array_downsample)):
    plt.plot(y_array, f_xy_zz[:, (int(i) - 1) * down_step],  color=colors[int(i) - 1])

plt.axhline(y=0, color="k", linestyle="-")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.xlabel("$x_1$")
plt.ylabel("%f(x_1, x_2=b)%")
ax.set_xlim(xx.min(), xx.max())
plt.show()
```

## 10.6 等高线：由函数值相等点连成

### 等高线

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_surface(xx, yy, f_xy_zz, cmap=cm.RdYlBu_r, rstride=5, cstride=5, linewidth=0.25, edgecolors=[0.5, 0.5, 0.5])
ax.set_proj_type("ortho")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$f(x,y)$")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.view_init(azim=-135, elev=30)
plt.tight_layout()
ax.grid(False)
plt.show()
```

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

ax.plot_wireframe(xx, yy, f_xy_zz, color=[0.5, 0.5, 0.5], linewidth=0.25)

colorbar = ax.contour(xx, yy, f_xy_zz, 20, cmap="RdYlBu_r")
fig.colorbar(colorbar, ax=ax)

ax.set_proj_type("ortho")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$f(x,y)$")
ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.view_init(azim=-135, elev=30)
plt.tight_layout()
ax.grid(False)
plt.show()
```

```python
fig, ax = plt.subplots()
colorbar = plt.contour(xx, yy, f_xy_zz, 20, cmap="RdYlBu_r")
fig.colorbar(colorbar, ax=ax)

ax.set_xlim(xx.min(), xx.max())
ax.set_ylim(yy.min(), yy.max())
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
plt.gca().set_aspect("equal", adjustable="box")
plt.show()
```

### 填充等高线
