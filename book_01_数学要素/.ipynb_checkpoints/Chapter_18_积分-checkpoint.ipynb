## 18.3 一元函数积分

### 导数、偏导、积分、二重积分

<font color="blue">积分</font>是微分的逆运算，积分关注变化累积。导数、微分、积分这个数学工具合称微积分，微积分是定量研究变化过程的重要数学工具。

### 一元函数积分

一元函数 $f(x)$ 自变量 $x$ 在区间 $[a,b]$ 上定积分运算记做:
$$
\int^{b}_{a}f(x)dx
$$
其中，$a$ 叫<font color="blue">积分下限</font>(lower bound)，$b$ 叫<font color="blue">积分上限</font>(upper bound)。

<font color="red">积分有正负之分，也就是说面积值有正负。</font>

对于一元函数，曲线在横轴之上包围的面积为正，曲线在横轴之下包围的面积为负。


```python
import numpy as np
from sympy import *
from matplotlib import pyplot as plt
```

```python
x = Symbol("x")
```

```python
# 定义函数和积分
f_x = x ** 2 + 1 / 2
f_x
```

```python
f_x_fcn = lambdify(x, f_x)
```

```python
integral_f_x = integrate(f_x, x)  # 计算 f(x) 的不定积分
integral_f_x
```

```python
integral_f_x_fcn = lambdify(x, integral_f_x)
```

```python
a = 0 # 积分下限
b = 1 # 积分上限

num = 201
x_array = np.linspace(-0.2, 1.2, num) # x 的取值范围
x_a_b_array = np.linspace(a, b, num) # 积分区间内的 x 取值范围
y_array = f_x_fcn(x_array)
y_a_b_array = f_x_fcn(x_a_b_array)
```

```python
# 计算定积分
integral_a_b = integral_f_x_fcn(b) - integral_f_x_fcn(a)
integral_a_b
```

```python
integral_a_b_v2 = integrate(f_x, (x, a, b))
integral_a_b_v2 = float(integral_a_b_v2)
integral_a_b_v2
```

```python
# 绘制函数和积分区间
fig, ax = plt.subplots()
ax.plot(x_array, y_array, "b")
ax.axvline(x=a, color="r", linestyle="-")
ax.axvline(x=b, color="r", linestyle="-")
ax.axhline(y=0, color="k", linestyle="-")

ax.fill_between(x_a_b_array, y_a_b_array, edgecolor="none", facecolor="#DBEEF3")

ax.set_xlim(-0.2, 1.2)
ax.set_ylim(-1, 2)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.set_title(r"$\int_a^b f(x)=%0.4f$" % integral_a_b)
```

```python
f_x_fcn = lambdify(x, f_x)
```

```python
integral_f_x = integrate(f_x, x)  # 计算 f(x) 的不定积分
integral_f_x
```

```python
integral_f_x_fcn = lambdify(x, integral_f_x)
```

```python
# 计算定积分
integral_a_b = integral_f_x_fcn(b) - integral_f_x_fcn(a)
integral_a_b
```

```python
integral_a_b_v2 = integrate(f_x, (x, a, b))
integral_a_b_v2 = float(integral_a_b_v2)
integral_a_b_v2
```

## 18.4 高斯函数积分

对于一元高斯函数积分，有如下积分结果:
$$
\int^{\infty}_{-\infty} \exp(-x^2)dx = \sqrt \pi
$$


## 定积分

积分公式：
$$
\int^{\infty}_{-\infty}\frac{1}{\sqrt2\pi}\exp \left(-\frac{x^2}{2}\right)dx = 1
$$

```python
import numpy as np
from sympy import *
import matplotlib.pyplot as plt
```

```python
x = Symbol("x")
```

```python
f_x = exp(- x ** 2)
f_x
```

```python
integrate(f_x, (x, -oo, oo))
```

```python
integrate(exp(-x**2/2), (x, -oo, oo))
```

```python
f_x_fcn = lambdify(x, f_x)
```

```python
integral_f_x = integrate(f_x, x)
integral_f_x
```

```python
integral_f_x_fcn = lambdify(x, integral_f_x)
```

```python
a = -0.5 # 积分下限
b = 1 # 积分上限

num = 201
x_array = np.linspace(-3, 3, num)
x_a_b_array = np.linspace(a, b, num)
y_array = f_x_fcn(x_array)
y_a_b_array = f_x_fcn(x_a_b_array)
```

```python
# 计算定积分
integral_a_b = integral_f_x_fcn(b) - integral_f_x_fcn(a)
integral_a_b
```

```python
integral_a_b_v2 = integrate(f_x, (x, a, b))
integral_a_b_v2 = float(integral_a_b_v2)
integral_a_b_v2
```

```python
# 绘制函数和积分区间
fig, ax = plt.subplots()
ax.plot(x_array, y_array, "b")
ax.axvline(x=a, color="r", linestyle="-")
ax.axvline(x=b, color="r", linestyle="-")
ax.axhline(y=0, color="k", linestyle="-")

ax.fill_between(x_a_b_array, y_a_b_array, edgecolor="none", facecolor="#DBEEF3")

ax.set_xlim(x_array.min(), x_array.max())
ax.set_ylim(np.floor(y_array.min()), np.ceil(y_array.max()))
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
ax.set_xlabel("x")
ax.set_ylabel("f(x)")
ax.set_title(r"$\int_a^b f(x)=%0.4f$" % integral_a_b)
```

```python
# 绘制累计积分函数
t = Symbol("t")
integral_f_x_oo_t = integrate(f_x, (x, -oo, t))
integral_f_x_oo_t_fcn = lambdify(t, integral_f_x_oo_t)

t_array = np.linspace(-3, 3, num)

integral_f_x_oo_t_array = integral_f_x_oo_t_fcn(t_array)

fig, ax = plt.subplots()
ax.plot(t_array, integral_f_x_oo_t_array, "b")
ax.axvline(x=a, color="r", linestyle="-")
ax.axvline(x=b, color="r", linestyle="-")
ax.axhline(y=integral_f_x_oo_t_fcn(a), color="r", linestyle="-")
ax.axhline(y=integral_f_x_oo_t_fcn(b), color="r", linestyle="-")

ax.set_xlim(t_array.min(), t_array.max())
ax.set_ylim(np.floor(integral_f_x_oo_t_array.min()), np.ceil(integral_f_x_oo_t_array.max()))
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
ax.set_xlabel("x")
ax.set_ylabel("Integral, F(x)")
ax.grid(linestyle="--", linewidth=0.25, color=[0.75, 0.75, 0.75])
```

## 18.5 误差函数：S 型函数的一种

$erf(x)$ 函数就是鼎鼎有名的误差函数(error function):
$$
\operatorname{erf}(x)
=\frac{1}{\sqrt{\pi}}\int_{-x}^{x}\exp(-t^2)\,dt
=\frac{2}{\sqrt{\pi}}\int_{0}^{x}\exp(-t^2)\,dt
$$
一般情况，误差函数自变量 $x$ 的取值为正值，但是为了计算方便，$erf()$ 的输入也可以是负值。$x$ 为负值时，下式成立:
$$
erf(x)=-erf(-x)
$$

```python
from sympy import oo, erf, lambdify
import numpy as np
from sympy.abc import x
from matplotlib import pyplot as plt
```

```python
x_array = np.linspace(-3, 3, 100)

erf_x_fcn = lambdify(x, erf(x))
y_array = erf_x_fcn(x_array)
```

```python
fig, ax = plt.subplots()
ax.plot(x_array, y_array, "b")

ax.axhline(y=erf(oo), color="r", linestyle="-")
ax.axhline(y=erf(-oo), color="r", linestyle="-")
ax.axhline(y=erf(0), color="r", linestyle="-")

ax.set_xlim(x_array.min(), x_array.max())
ax.set_ylim(-1, 1)
plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)
ax.set_xlabel("x")
ax.set_ylabel("erf(x)")
ax.grid(linestyle="--", linewidth=0.25, color=[0.75, 0.75, 0.75])
```

## 18.6 二重积分：类似二重求和

### 先对 $x$ 积分 

给定积分区域 $D=\{(x,y)|a<x<b,c<y<d\}$，$f(x,y)$ 二重积分记做:
$$
\int_c^d \int_a^b f(x,y)\,dx\,dy
$$

请注意上式二重积分的先后次序，先对 $x$ 积分，再对 $y$ 积分。也就是说，内部这一层 $\int^{x=b}_{x=a}(x,y)dx$ 先消去 $x$，变成有关 $y$ 的一元函数，然后再对 $y$ 积分:
$$
\int_c^d \int_a^b f(x,y)\,dx\,dy=
\int_{y=c}^{y=d}
\underbrace{
\overbrace{
\int_{x=a}^{x=b} f(x,y)\,dx
}^{\text{Eliminate }x}
}_{\text{A function of }y}
\,dy
$$

从几何角度，当 $y=c$ 时，$\int^{xx=b}_{x=a}(x,y=c)dx$ 结果为图中暖色阴影区域面积，也就是压缩为一个值。


### 先对 $y$ 积分

先对 $y$ 积分，$\int^{y=d}_{y=c}(x,y)dy$相当于消去 $y$，得到有关 $x$ 的一元函数，然后再对 $x$ 积分:
$$
\int_a^b \int_c^d f(x,y)\,dy\,dx=
\int_{x=a}^{x=b}
\underbrace{
\overbrace{
\int_{y=c}^{y=d} f(x,y)\,dy
}^{\text{Eliminate }y}
}_{\text{A function of }x}
\,dx
$$

当 $x=a$ 时，$\int^{y=d}_{y=c}(x=a,y)dy$ 结果为图中冷色阴影区域面积，即压缩为一个值。


### 调换积分顺序

<font color="red">二重积分、多重积分中，积分先后顺序不能随意调换。</font>

### 二元高斯函数

```python
from sympy.abc import x, y, s, t
from sympy import *
```

举个例子，给定二元高斯函数 $f(x,y)$，
$$
f(x, y) = \exp(-x^2-y^2)
$$

```python
f_xy = exp(-x**2 - y**2)
f_xy
```

$f(x,y)$ 的二重不定积分 $F(x,y)$ 可以用误差函数表达为:
$$
F(x,y)=
\int_{-\infty}^{y}\int_{-\infty}^{x}
\exp(-u^2-v^2)\,du\,dv=
\frac{\pi}{4}\operatorname{erf}(x)\operatorname{erf}(y)
+\frac{\pi}{4}\operatorname{erf}(x)
+\frac{\pi}{4}\operatorname{erf}(y)
+\frac{\pi}{4}
$$

```python
f_x_y_double_integrate = integrate(f_xy, (y, -oo, y), (x, -oo, x))
f_x_y_double_integrate
```

特别地，$f(x,y)$ 曲面和整个水平面围成的体积为 $π$，即，
$$
\int_{-\infty}^{\infty}\int_{-\infty}^{\infty}
\exp(-x^2-y^2)\,dx\,dy
=\pi
$$

```python
f_x_y_volume = integrate(f_xy, (y, -oo, oo), (x, -oo, oo))
f_x_y_volume
```

## 18.7 "偏积分"：类似偏求和

```python
from sympy.abc import x, y, s, t
from sympy import *
```

本节自创一个积分概念——“偏积分”，如下两式就是“偏积分”:
$$
\int^{b}_{a}f(x, y)dx
$$
$$
\int^{d}_{c}f(x, y)dy
$$

```python
f_xy = exp(- x ** 2 - y ** 2)
f_xy
```

### 对 $y$ 偏积分

给定二元高斯函数 $f(x, y)$:
$$
f(x,y) = \exp(-x^2-y^2)
$$
$f(x,y)$ 对于 $y$ 从负无穷到正无穷偏积分，得到结果变成了关于 $x$ 的高斯函数:
$$
\int_{-\infty}^{\infty}\exp(-x^2-y^2)\,dy=
\sqrt{\pi}\exp(-x^2)
$$



```python
f_x_partial_integrate = integrate(f_xy, (y, -oo, oo))
f_x_partial_integrate
```

### 对 $x$ 偏积分

类似地，二元高斯函数 $f(x,y)$ 对 $x$ 从负无穷到正无穷偏积分，结果为关于 $y$ 的高斯函数:
$$
\int_{-\infty}^{\infty}\exp(-x^2-y^2)dx=
\sqrt{\pi}\exp(-y^2)
$$


```python
f_y_partial_integrate = integrate(f_xy, (x, -oo, oo))
f_y_partial_integrate
```

## 18.8 估算圆周率：牛顿法

于是圆周率可以通过下式近似得到:
$$
\pi=24\times\left(
\frac{\sqrt{3}}{32}-
\sum_{n=0}^{\infty}
\frac{(2n)!}
{2^{4n+2}(n!)^2(2n-1)(2n+3)}
\right)
$$
圆周率估算结果随 $n$ 增加而不断收敛。观察曲线，可以发现这个估算过程收敛的速度很快。

```python
import numpy as np
from scipy.special import factorial
import matplotlib.pyplot as plt
```

```python
n_array = np.linspace(0, 10, 11) 
# 计算级数展开项
expansion = factorial(2 * n_array) / 2**(4 * n_array + 2) / (factorial(n_array))**2 / (2 * n_array - 1) / (2 * n_array + 3)
expansion
```

```python
# 计算 π 的估计值
esp_pi = 24 * (np.sqrt(3) / 32 - np.cumsum(expansion))
esp_pi
```

```python
# 绘制 π 的估计值图像
fig, ax = plt.subplots()

plt.axhline(y=np.pi, color="red", linestyle="-")
plt.plot(n_array, esp_pi, color="b", marker="x")

plt.tight_layout()
plt.xlabel("n")
plt.ylabel(r"Estimate of $\pi$")
```

### 拓展：蒙特卡洛模拟

<font color="blue">蒙特卡洛模拟</font>(Monte Carlo simulation)在大数据分析和机器学习中占据重要的位置。蒙特卡洛模拟以摩纳哥的赌城蒙特卡罗命名，是一种使用随机数并以概率理论为指导的数值计算方法。

在单位圆 $(r=1)$ 的周围，构造一个以圆心为中心、以圆直径为边长的外切正方形。圆形面积 $A_{circle}$ 和正方形面积 $A_{square}$ 容易求得:
$$
\begin{cases}
A_{\text{circle}}=\pi r^2=\pi\\
A_{\text{square}}=(2r)^2=4
\end{cases}
$$

进而求得圆周率 $π$ 和两个面积的比例关系:
$$
\pi = 4×\frac{A_{circle}}{A_{square}}
$$

统计落入圆内的数据点个数 $m$ 与总数据点总数 $n$的比值，这个比值即为圆面积和正方形面积之比近似值。


这种估算圆周率的方法思想来源于在18世纪提出的<font color="blue">布丰投针问题</font>(Buffon'sneedle problem)。

## 18.9 数值积分：黎曼求积

### 将平面图形切成细长条

德国数学家黎曼(Bernhard Riemann,1826~1866)提出了一个求积解决方案——将不规则图形切成细长条。然后这些细长条近似看成一个个矩形，计算出它们的面积。这些矩形面积求和，可以用来近似不规则形状的面积。

狭长长方形越细，也就是 $\Delta x$ 越小，长方形越贴合区域形状，就越能精确估算面积。特别地，当细长条的宽度 $\Delta x$ 趋近于 $0$ 时，得到的面积的极限值就是不规则形状的面积。



### 将立体图形切成细高立方体

我们可以用一个个细高立方体体积之和来近似估算几何体的体积。

随着细长立方体不断变小，这些立方体的体积之和不断接近不规则几何体的体积。


### 基本数值积分方法

```python
import numpy as np
import matplotlib.pyplot as plt
from sympy import *
```

```python
x = Symbol("x")
```

```python
f_x = x ** 2
f_x_fcn = lambdify(x, f_x)
f_x
```

```python
# 计算不定积分
integral_f_x = integrate(f_x, x)
integral_f_x
```

```python
integral_f_x_fcn = lambdify(x, integral_f_x)
```

```python
# 定义上下限
a = 0
b = 1
# # 使用不定积分计算定积分值
integral_a_b = integral_f_x_fcn(b) - integral_f_x_fcn(a)
integral_a_b_v2 = integrate(f_x, (x, a, b))
integral_a_b_v2 = float(integral_a_b_v2)
integral_a_b_v2
```

```python
num_interval = 20 # 将区间划分为 20 个小区间
delta_x = (b - a) / num_interval

x_array = np.linspace(a, b , num_interval + 1)
y_array = f_x_fcn(x_array)

x_array_fine = np.linspace(a, b, 200)
y_array_fine = f_x_fcn(x_array_fine)
```

在 $[a,b]$ 区间内，为估算函数在区间内和 $x$ 轴形成的面积，用左侧 $a$ 点的函数值 $f(a)$ 进行积分估值运算， 这种方法叫做向前差分，也叫left Riemann sum:
$$
\int_{a}^{b}f(x)dx \approx (b-a)f(a)
$$

```python
# 左 Riemann 和
fig = plt.figure(figsize=(5, 5))
ax = fig.add_subplot(1,1,1)

ax.plot(x_array_fine, y_array_fine, color="#0070C0")

x_left = x_array[:-1]
y_left = y_array[:-1]

plt.plot(x_left, y_left, "rx", markersize=10)

plt.bar(x_left, y_left, width=delta_x, facecolor="#DEEAF6", algin="edge", edgecolor="#B2B2B2")

ax.axvline(x=a, color="r", linestyle="-")
ax.axvline(x=b, color="r", linestyle="-")

left_riemann_sum = np.sum(f_x_fcn(x_left), * deta_x)

plt.title('Left Riemann sum (N = %0.0f) = %0.3f'
          % (num_interval, left_riemann_sum)) 

plt.xlim((a, b))
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['top'].set_visible(False)
plt.xlabel('x')
plt.ylabel('f(x)')
```

用 $[a,b]$ 区间右侧 $b$ 点函数值 $f(b)$ 进行积分估值运算叫做向后差分，也叫right Riemann sum:
$$
\int_{a}^{b}f(x)dx \approx (b-a)f(b)
$$

用 $[a,b]$ 区间中间点 $(a+b)/2$ 函数值 $f((a+b)/2)$ 作为矩形高度来估值叫做中值差分，也叫 middle Riemann sum:
$$
\int_{a}^{b}f(x)dx \approx (b-a)f\left(\frac{a+b}{2}\right)
$$

梯形法，这种方法用 $f(a)$ 和 $f(b)$ 的平均值:
$$
\int_{a}^{b}f(x)dx \approx (b-a)\left(\frac{f(a)+f(b)}{2}\right)
$$
