# 12 超越函数(Transcendental Functions) 

## 12.1 指数函数：指数为自变量

<font color="blue">指数函数</font>(exponentialfunction)的一般形式为:
$$
f(x)=b^x
$$
其中，$b$ 为<font color="blue">底数</font>(base)，自变量 $x$ 为<font color="blue">指数</font>(exponent)。

$b>1$ 时，$f(x)=b^x$ 单调递增；$0<b<1$ 时，$f(x)=b^x$ 单调递减。


### 自然指数函数

指数函数指的是<font color="blue">自然指数函数</font>(natural exponential function):
$$
f(x)=e^x=exp(x)
$$
自然指数函数中的“自然”指的是自然常数 $e$ 为底数，$e≈2.718$。

### 指数增长

<font color="blue">指数增长</font>(exponential growth)模型就是用如下指数函数来表达:
$$
G(t)=(1+r)^t
$$
$r$ 为年化增长率，$t$ 是年限。

## 12.2 对数函数：把连乘变成连加

<font color="blue">对数函数</font>(logarithmic function)解析式如下:
$$
y=f(x)=log_b(x)
$$

其中，$b$ 为<font color="blue">对数底数</font>(logarithmic base)，$b>0$ 且 $b≠1$。上述对数函数的定义域为 $x>0$。

### 自然对象函数

<font color="blue">自然对数函数</font>(natural logarithmic function)如下:
$$
y=f(x)=ln(x)=log_e(x)
$$



### 对数运算特点

$$
\log_bx = \frac{\log_kx}{\log_kb}
$$
$$
\log_bx = \frac{\log_{10}x}{\log_{10}b} = \frac{\ln x}{\ln x}
$$
$$
\log_{b^n}x^m = \frac{m}{n}\log_bx
$$
$$
x=b^{\log_b(x)}
$$
$$
x^{\log_b(y)}=y^{\log_b(x)}
$$

对数的一个重要的性质是，把连乘变成连加：
$$
\log_b(xyz)=\log_bx+\log_by+log_bz
$$

### 对数刻度

<font color="blue">对数刻度</font>(logarithmic scale或logarithmic axis)是一种<font color="blue">非线性刻度</font>(nonlinear scale)，常用来描述较大的数值。



```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p = plt.rcParams
plt.rcParams["font.sans-serif"] = ["Microsoft YaHei"]
p["ytick.minor.visible"] = True
p["xtick.minor.visible"] = True
p["axes.grid"] = True
p["grid.color"] = "0.5"
p["grid.linewidth"] = 0.5
```

```python
f1 = 10 ** x1
f2 = x1
f3 = np.log(x)

fig, ax = plt.subplots()
plt.plot(x1, f1, color="r")
plt.plot(x1, f2, color="k")
plt.plot(x, f3, color="b")
plt.xlim(-1, 10)
plt.ylim(-1, 10)
ax.grid(which="major", linestyle="--", linewidth=0.5, color="0.5")
ax.grid(which="minor", linestyle="--", linewidth=0.3, color="0.8")
plt.tight_layout()
ax.set_box_aspect(1)
```

## 12.3 高斯函数：高斯分布之基础

### 高斯函数

自然指数函数复合二次函数，得到<font color="blue">高斯函数</font>(Gaussian function):
$$
f(x)=\exp(-\gamma x^2)
$$
其中，$\gamma$ 为参数，$\gamma>0$。高斯函数的定义域是 $(-∞,+∞)$，而值域是   $(0,1]$。给出的高斯函数关于 $x=0$ 对称。

### 高斯小传

<font color="blue">略</font>

### 拉普拉斯核函数

此外，绝对值函数 $|x|$和指数函数的复合，得到的是一元<font color="blue">拉普拉斯核函数</font>(Laplacian kernel function):
$$
f(x)=\exp(-\gamma |x|)
$$

## 12.4 逻辑函数：在 0 和 1 直接取值

<font color="blue">逻辑函数</font>(logistic function)也可以视作是自然指数函数扩展得到的复合函数。下式为最简单的一元逻辑函数:
$$
f(x)=\frac{1}{1+\exp(-x)} = \frac{\exp(x)}{1+\exp(x)}
$$

更一般的一元逻辑函数形式为:
$$
f(x)=\frac{1}{1+\exp(-(b_0+b_1x))}
$$

### 中心点位置

$x=-\frac{b_0}{b_1}$ 这个点被称作为逻辑函数中心所在位置。$b_0$ 决定逻辑函数中心所在位置。

### 逻辑斯谛增长

逻辑曲线可以用来模拟人口增长的<font color="blue">S 型增长曲线</font>(S-shaped growth curve).



### S 型曲线

逻辑函数是<font color="blue">S 型函数</font>(sigmoid function 或 S-shaped function)的一种。

常见的 S 型函数还有双曲正切函数$f(x)=tanh(x)$、反正切函数$f(x)=arctan(x)$、误差函数$f(x)=erf (x)$等。

### tanh()函数

sigmoid 函数特指 tanh() 函数。给定如下 tanh() 函数一般式；
$$
f(x) = tanh(\gamma x)
$$

## 12.6 函数变换：平移、缩放、对称

```python
import matplotlib.pyplot as plt
import numpy as np
from sympy.abc import x
from sympy import exp, lambdify
```

```python
def plot_curve(x_array, y_array,
               x_array_new, y_array_new): # 变换后曲线函数的参数
    fig, ax = plt.subplots()

    plt.plot(x_array, y_array, color="#0070c0", label="Origianl")
    ax.fill_between(x_array, y_array, edgecolor="none", facecolor="#0070c0", alpha=0.2)

    plt.plot(x_array_new, y_array_new, color="r", label="Transformed")

    plt.xlabel("x")
    plt.ylabel("y")
    plt.axhline(y=0, color="k", linestyle="-")
    plt.axvline(x=0, color="k", linestyle="-")
    plt.xticks(np.arange(-4, 5, step=1))
    plt.yticks(np.arange(-4, 5, step=1))
    plt.axis("scaled")

    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(False)

    plt.legend()
    ax.grid(color=[0.5, 0.5, 0.5], linewidth=0.25, linestyle="--")
    plt.rcParams["figure.figsize"] = [7.50, 3.50]
    plt.rcParams["figure.autolayout"] = True
```

```python
x_array = np.arange(-4, 4 + 0.01, step=0.01)

f_x  = 2 * exp(-(x-1) ** 2)
f_x_fcn = lambdify([x], f_x)
f_x_array = f_x_fcn(x_array)
```

### 平移

相对 $y=f(x)$，$f(x)+c$ 竖直向上平移 $c$ 单位(vertical shiftup bycunits)；$f(x)-c$ 将原函数竖直向下平移 $c$ 单位(vertical shift down bycunits)。

```python
for c in [2, -3]:
    f_x_array_new = f_x_array + c
    plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

相对 $y=f(x)$，$f(x+c)$ 相当于函数向左平移 $c$ 单位(horizontal shift left bycunits),$ c>0$ ；$f(x-c)$ 相对原函数向右平移 $c$ 单位(horizontal shift right by c units).

```python
for c in [3, -1]:
    f_x_new = 2 * exp(-((x+c)-1) ** 2)
    f_x_new_fcn = lambdify([x], f_x_new)
    f_x_array_new = f_x_new_fcn(x_array)
    plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

### 缩放

相对 $y=f(x)$，$cf(x)$ 相当于函数<font color="blue">竖直方向缩放</font>(vertical scaling)。$c>1$ 时，$cf(x)$ <font color="blue">竖直方向拉伸</font>(vertical stretch)；$0<c<1$ 时，$cf(x)$ <font color="blue">竖直方向压缩</font>(vertical compression).

```python
for c in [1/2, 2]:
    f_x_array_new = c * f_x_array
    plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

相对 $y=f(x)$, $f(cx)$ 相当于函数<font color="blue">水平方向伸缩</font>(horizontal scaling)。$c>1$ 时，<font color="blue">水平方向压缩</font>(horizontal compression)；$0<c<1$ 时，<font color="blue">水平方向拉伸</font>(horizontal stretch)。面积等比例缩放，缩放比例为 $1/c$。

```python
for c in [1/2, 2]:
    f_x_new = 2 * exp(-((c * x)-1) ** 2)
    f_x_new_fcn = lambdify([x], f_x_new)
    f_x_array_new = f_x_new_fcn(x_array)
    plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

### 对称

相对 $y=f(x)$，$f(-x)$ 相当于函数关于 $y$ 轴对称(reflectionabout yaxis)。$-f(x)$ 相当于函数关于 $x$ 轴对称(reflectionaboutxaxis)。而 $f(x)$ 和 $-f(-x)$ 关于原点对称。

```python
f_x_new = 2 * exp(-((-x)-1)**2)
f_x_new_fcn = lambdify([x], f_x_new)
f_x_array_new = f_x_new_fcn(x_array)
plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

```python
f_x_array_new = - f_x_array
plot_curve(x_array, f_x_array, x_array, f_x_array_new)
```

### 拓展

一元高斯分布的概率密度函数解析式如下:
$$
p(x)=\frac{1}{\sigma\sqrt{2\pi}}\exp \left(\frac{-1}{2}\left(\frac{x-\mu}{\sigma}\right)^2\right)
$$

所谓概率密度函数上的点并不是概率，概率是面积，是针对与概率密度函数的积分。
