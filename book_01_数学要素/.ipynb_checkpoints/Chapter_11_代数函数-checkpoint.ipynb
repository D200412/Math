## 11.2 一次函数：一条斜线

### 四种形式

<font color="blue">一次函数</font>(linear function)有几种不同的形式构造:
- <font color="blue">斜截式</font>(slope-intercept form);
- <font color="blue">点斜式</font>(point-slope form);
- <font color="blue">两点式</font>(two-point form);
- <font color="blue">参数方程</font>(parametric equation)。

### 斜截式

斜截式一次函数形式如下:
$$
y=f(x)=ax+b
$$
斜截式需要两个参数——斜率 $(a)$ 和 $y$ 轴截距 $(b)$。

  当 $a=0$ 时，为<font color="blue">常数函数</font>(constant function)。也就是说，<font color="blue">零斜率</font>(zero slope)对应常数函数，即<font color="blue">水平线</font>(horizontal line)。

  <font color="blue">无定义斜率</font>(undefined slope)代表一条<font color="blue">竖直线</font>(vertical line)，此时图像虽然是一条直线，垂直于横轴，但它并不是函数。

  当 $b=0$ 时，函数为<font color="blue">比例函数</font>(proportional function)，而 $a$ 则叫做<font color="blue">比例常数</font>(constant ratio或proportionality constant)。比例函数是特殊的一次函数。

一次函数有两种斜率：<font color="blue">正斜率</font>(positive slope)和<font color="blue">负斜率</font>(negative slope)。一次函数斜率大于 0，单调递增。斜率小于 0，一次函数单调递减。

```python
import numpy as np
import matplotlib.pyplot as plt

def plot_lines(x_array, y_array, labels, title, leg=2):
    fig, ax = plt.subplots()
    colors = plt.cm.jet(np.linspace(0, 1, len(y_array)))

    for i, (y, label) in enumerate(zip(y_array, labels)):
        ax.plot(x_array, y, label=label, color=colors[i])

    ax.set_title(title)
    ax.set_xlabel("$x$")
    ax.set_ylabel("$f(x)$")
    plt.legend()
    plt.xticks(np.arange(-leg, leg + .5, step=0.5))
    plt.yticks(np.arange(-leg, leg + .5, step=0.5))
    plt.axis("scaled")
    ax.set_xlim(-leg, leg)
    ax.set_ylim(-leg, leg)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.grid(linestyle="--", color=[0.5, 0.5, 0.5], linewidth=0.25)
```

### 斜率 

```python
w_array = np.array([1/5, 1/4, 1/3, 1/2, 1, 2, 3, 4, 5])
x_array = np.linspace(-2, 2, 100)
y_array = [w * x_array for w in w_array]
labels = [f"w1 = {w:.2f}" for w in w_array]

plot_lines(x_array, y_array, labels, "y = w1 * x (w1 > 0)")
```

```python
w_array_negative = -w_array
y_array_negative = [w * x_array for w in w_array_negative]
labels_negative = [f"w1 = {w:.2f}" for w in w_array]

plot_lines(x_array, y_array_negative, labels_negative, "y = w1 * x (w1 < 0)")
```

### 截距

```python
x_array = np.linspace(-4, 4, 100)
b_array = np.arange(-3, 4)
y_arrays_b = [x_array + b for b in b_array]
labels_b = [f"b = {b}" for b in b_array]

plot_lines(x_array, y_arrays_b, labels_b, "y = x + b", leg=4)
```

```python
y_arrays_neg_x_b = [-x_array + b for b in b_array]
labels_neg_x_b = [f"b = {b}" for b in b_array]

plot_lines(x_array, y_arrays_neg_x_b, labels_neg_x_b, "y = -x + b", leg=4)
```

### 两条直线关系

如果两条直线斜率相同，它们相互<font color="blue">平行</font>(parallelize)或<font color="blue">重合</font>(coincide)，否则两条直线<font color="blue">相交</font>(intersect)，有唯一交点。如果两个一次函数的斜率乘积为 -1，则两者<font color="blue">垂直</font>(perpendicular)。



### 点斜式

一次函数的第二种是点斜式：
$$
y-y^{(1)} = f(x)-y^{(1)}=a(x-x^{(1)})
$$

### 两点式

第三种形式是两点式，即两点确定一条直线。
$$
y-y^{(1)}=\frac{y^{(2)} - y^{(1)}}{x^{(2)} - x^{(1)}}(x-x^{(1)})
$$
其中，$x^{(1)} \ne x^{(2)}$

## 11.3 二次函数：一条抛物线

<font color="blue">二次函数</font>(quadratic function)是<font color="blue">二次多项式函数</font>(second order polynomial function 或 second degree polynomial function).

二次函数图象是<font color="blue">抛物线</font>(parabola)，可以<font color="blue">开口向上</font>(open upward)或<font color="blue">开口向下</font>(opendownward)，<font color="blue">对称轴平行于纵轴</font>(the axis of symmetry is parallel to the y-axis)。

### 三种形式

二次函数解析式有三种形式:
- <font color="blue">基本式</font>(standard form)；
- <font color="blue">两根式</font>(factored form)；
- <font color="blue">顶点式</font>(vertexform)。

### 基本式

$$
f(x)=ax^2+bx+c,\; a≠0
$$

### 开口大小

二次函数图像开口大小随系数 $a$ 变化。$a$ 的绝对值越大，开口越小，对应二次函数图像变化越剧烈。

### 两根式

如果 $f(x)=0$ 存在两个实数根的话，二次函数可以写成两根式：
$$
f(x)=a(x-r_1)(x-r_2), \; a≠0
$$
$r_1$ 和 $r_2$ 为二次方程的根

### 顶点式

二次函数另外一个常见的形式是顶点式，具体形式如下：
$$
f(x)=a(x-h)^2+k,\; a≠0
$$
其中，$h$ 和 $k$ 分别为顶点的横纵坐标值。

$h$ 影响函数在水平方向位置，$k$ 影响函数在竖直方向位置。

```python
import numpy as np
import matplotlib.pyplot as plt

def drawing(coefficient, variable, result, y_range, label, step=0.5):
    # coefficient: 系数数组
    # variable: x的范围
    # result: 函数值
    # y_range: y的范围
    fig, ax = plt.subplots()
    colors = plt.cm.jet(np.linspace(0, 1, result.shape[1]))
    
    for i in range(1, result.shape[1] + 1):
        plt.plot(variable, result[:, i-1], color=colors[i-1], label=f"${label} = {coefficient[i-1]:.0f}$")
    
    plt.xlabel("$x$")
    plt.ylabel("$f(x)$", rotation=0)
    plt.legend()
    plt.axis("scaled")
    plt.xticks(np.arange(variable.min(), variable.max()+step, step=step))
    plt.yticks(np.arange(y_range[0], y_range[1]+step, step=step))
    plt.xlim(variable.min(), variable.max())
    plt.ylim(y_range[0], y_range[1])
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.grid(linestyle="--", color=[0.5, 0.5, 0.5], linewidth=0.25)
```

```python
a_array = np.linspace(-6, -1, 6)
x_array = np.linspace(-2, 2, 100)
aa, xx = np.meshgrid(a_array, x_array)

ww = aa * xx ** 2
y_range = [np.min(ww, axis=0).max(), 0]

drawing(a_array, x_array, ww, y_range, "a")
```

```python
a_array = np.linspace(1, 6, 6)
x_array = np.linspace(-2, 2, 100)
aa, xx = np.meshgrid(a_array, x_array)

ww = aa * xx ** 2
y_range = [0, np.max(ww, axis=0).min()]

drawing(a_array, x_array, ww, y_range, "a")
```

```python
h_array = np.linspace(-4, 4, 9)
x_array = np.linspace(-5, 5, 100)

hh, xx = np.meshgrid(h_array, x_array)

ww = 1 * (xx - hh) ** 2 + 0

y_range = [0, 10]

drawing(h_array, x_array, ww, y_range, "h", step=1)
```

```python
k_array = np.linspace(-4, 4, 9)
x_array = np.linspace(-5, 5, 100)

kk, xx = np.meshgrid(k_array, x_array)

ww = 1 * (xx - 0) ** 2 + kk

y_range = [-4.5, 4.5]

drawing(k_array, x_array, ww, y_range, "k", step=1)
```

## 11.4 多项式函数：从叠加角度看

<font color="blue">多项式函数</font>(polynomial function)相当于一次和二次函数的推广，具体形式如下：
$$
y=f(x)=a_Kx^K+a_{K-1}x^{K-1}+…+a_2x^2+a_1x+a_0=\sum^{K}_{i=0}a_ix^i
$$
其中，最高次项系数 $a_k$ 不为 0，$K$ 为最高次项次数。

### 三次函数

<font color="blue">三次函数</font>(cubic function, polynomial function of degree 3)的形式为：
$$
y=f(x)=a_3x^3+a_2x^2+a_1x+a_0
$$

```python
import math
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 100)

def plot_curve(x, y, xlim, ylim):
    fig, ax = plt.subplots()

    plt.xlabel(r"$\it{x}$")
    plt.ylabel(r"$\it{f}(\it{x})$")
    plt.plot(x, y, linewidth=1.5)
    plt.axhline(y=0, color="k", linewidth=1.5)
    plt.axvline(x=0, color="k", linewidth=1.5)
    ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
    ax.set_xticks(np.arange(-2, 2.5, step=0.5))
    ax.set_yticks(np.arange(y.min(), y.max()+0.5, step=0.5))
    ax.set_xlim(xlim)
    ax.set_ylim(ylim)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.set_aspect("equal", "box")
```

```python
# 线性函数 y = x + 1
y = x + 1
plot_curve(x, y, (-2, 2), (-1, 3))
```

```python
# 线性函数 y = -x + 1
y = -x + 1
plot_curve(x, y, (-2, 2), (-1, 3))
```

```python
# 二次函数，开口向上 y = x^2 - 2
y = x ** 2 - 2
plot_curve(x, y, (-2, 2), (-2, 2))
```

```python
# 二次函数，开口向下 y = -x^2 + 2
y = -x ** 2 + 2
plot_curve(x, y, (-2, 2), (-2, 2))
```

```python
# 三次函数 y = x^3 - x
y = x**3 - x
plot_curve(x, y, (-2, 2), (-2, 2))
```

```python
# 三次函数 y = -x^3 + x
y = -x**3 + x
plot_curve(x, y, (-2, 2), (-2, 2))
```

## 11.5 幂函数：底数为自变量

<font color="blue">幂函数</font>(power function)是形如下式的函数:
$$
f(x)=k·x^p
$$
自变量 $x$ 为<font color="blue">底数</font>(base)，$p$ 为<font color="blue">指数</font>(exponent 或 power)

### 平方根函数

<font color="blue">略</font>

### 立方根函数

<font color="blue">略</font>

### 奇偶性

当 $p$ 为偶数时，幂函数为偶函数，图像关于 $y$ 轴对称。$p$ 值越大，$x$ 绝对值增大时，函数值越快速接近正无穷。

当 $p$ 为奇数时，幂函数为奇函数，图像关于原点对称。$p$ 值越大，$x$ 绝对值增大时，函数值越快速接近正无穷或负无穷。

### 反比例函数

<font color="blue">反比例函数</font>(inversely proportional function)的一般式:
$$
y=f(x)=\frac{k}{x}
$$

和 $y=f(x)=frac{1}/{x}$ 相比，$|k|>1$时，双曲线朝远离原点方向拉伸；$|k|<1$ 将双曲线向靠近原点方向压缩。



### 渐近线

<font color="blue">略</font>

### 有理函数

反比例函数移动之后可以得到最简单的<font color="blue">有理函数</font>(rationalfunction)，解析式如下:
$$
f(x)=\frac{k}{x-h}+a
$$
其中，$x≠h$。

## 11.6 分段函数：不连续函数

<font color="blue">分段函数</font>(piecewise function)是一类不连续函数；分段函数是自变量 $x$ 的不同的取值范围有不同的解析式的函数。

<font color="red">注：分段函数不能算作代数函数</font>

### 绝对值函数

<font color="blue">绝对值函数</font>(absolute value function)可以看做是分段函数。绝对值函数的一般式为:
$$
f(x)=k|x-h|+a
$$
