## 22.1 向量：有大小、有方向

### 向量简史

<font color="blue">略</font>

### 向量的几何意义

<font color="blue">向量</font>(vector)<font color="blue">是既有长度又有方向的量</font>(a quantity that possesses both magnitude and direction)。物理学中，<font color="blue">位移</font>(displacement)、<font color="blue">速度</font>(velocity)、<font color="blue">加速度</font>(acceleration)、<font color="blue">力</font>(force)等物理量都是向量。

位移向量的大小代表前进的距离大小，而向量的方向代表位移方向。

和向量相对的是标量。<font color="blue">标量(scalar,scalar quantity)是有大小没有方向的量，用实数表示。

### 向量定义

给定一个列向量 $\mathit{a}$ :
$$
\boldsymbol{a}=
\begin{bmatrix}
4 \\
3
\end{bmatrix}
$$

向量 $\mathit{a}$ 可以表达为一个带箭头的线段，它以原点 $O(0,0)$ 为起点指向终点 $A(4,3)$.
因此，向量 $\mathit{a}$ 也可以写作 $\overrightarrow{OA}$:
$$
\overrightarrow{OA}
=\begin{bmatrix}
4 \\
3
\end{bmatrix}-
\begin{bmatrix}
0 \\
0
\end{bmatrix}
=\begin{bmatrix}
4 \\
3
\end{bmatrix}
$$


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 从原点绘制向量的函数
def draw_vector(vector, RBG, label):
    array = np.array([[0, 0, vector[0], vector[1]]])
    X, Y, U, V = zip(*array)
    plt.quiver(X, Y, U, V, angles="xy", scale_units="xy", scale=1, color=RBG)

    label = label + f"({vector[0]},{vector[1]})"
    plt.annotate(label, (vector[0], vector[1]), textcoords="offset points", xytext=(0, 10),
                 ha="center")
```

```python
# 定义向量
a = np.array([4, 3])
i = np.array([1, 0])
j = np.array([0, 1])
```

```python
# 可视化
fig, ax = plt.subplots()

draw_vector(4 * i, np.array([0, 112, 192])/255, "4i")
draw_vector(3 * j, np.array([255, 0, 0])/255, "3j")

draw_vector(i, np.array([0, 112, 192])/255, "i")
draw_vector(j, np.array([255, 0, 0])/255, "j")

draw_vector(a, np.array([146, 208, 80])/255, "a")

plt.xlabel("$x$")
plt.ylabel("$y$")

plt.axis("scaled")
ax.set_xlim([0, 5])
ax.set_ylim([0, 5])
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
plt.show()
```

### 向量模

向量 $\mathit{a}$ 的长度就是线段 $OA$ 的长度:
$$
\|\mathit{a}\| = \sqrt{4^2 + 3^2} = 5
$$
$\|\mathit{a}\|$ 计算向量 $mathit{a}$ 的长度，$\|\mathit{a}\|$ 叫向量模，也叫 $L^2$ 范数。$L^2$ 范数是 $L^P$ 范数的一种。

### 向量分解

类似物理学中力的分解，一个向量还可以表达成若干向量的和。

<font color="blue">单位向量</font>(unit vector)是指模(长度)等于 1 的向量。

### 向量模

另外一个角度，将 $A$ 和 $B$ 的坐标写成向量 $\mathit{a}$ 和 $\mathit{b}$ ，线段 $AB$ 长度等价于 $\mathit{a}-\mathit{b}$ 的模:
$$
d = \| \mathit{a}-\mathit{b}\| = \sqrt{\left(\mathit{a}-\mathit{b}\right)·\left(\mathit{a} - \mathit{b}\right)}
$$

## 22.4 向量内积和向量夹角

```python
import numpy as np
```

```python
a = [4, 1]
b = [1, 3]
```

```python
# 计算 L2 范数
a_norm = np.linalg.norm(a)
a_norm
```

```python
b_norm = np.linalg.norm(b)
b_norm
```

向量 $\boldsymbol{a}$ 和 $\boldsymbol{b}$ 内积为: 
$$
\boldsymbol{a}·\boldsymbol{b}=a_1b_1+a_2b_2+...+a_{D}b_{D}=\sum^{D}_{i=1}a_ib_i
$$
有了几何视角，向量 $\boldsymbol{a}$ 和 $\boldsymbol{b}$ 的内积有了一个新的定义方式:
$$
\boldsymbol{a} · \boldsymbol{b} = \|\boldsymbol{a}\|\|\boldsymbol{b}\|\cos\theta
$$

$\boldsymbol{a}$ 和 $\boldsymbol{b}$ 的内积为 $\boldsymbol{a}$ 的模乘向量 $\boldsymbol{b}$  在向 $\boldsymbol{a}$ 方向上投影的分量值。$\boldsymbol{b}$ 在 $\boldsymbol{a}$ 方向上投影的分量值为 $\|\boldsymbol{b}\|\cosθ$，这个值也叫 $\boldsymbol{b}$ 在 $\boldsymbol{a}$ 方向上的<font color="blue">标量投影</font>(scalar projection)。



```python
a_dot_b= np.dot(a, b)
a_dot_b
```

### 向量夹角

向量 $\boldsymbol{a}$ 和 $\boldsymbol{b}$ 的夹角 $θ$ 的余弦值可以通过下式求得:
$$
\cos\theta=\frac{\boldsymbol{a}·\boldsymbol{b}}{\|\boldsymbol{a}\|\|\boldsymbol{b}\|}
$$

```python
cos_result = a_dot_b / (a_norm * b_norm)
cos_result
```

```python
angle = np.degrees(np.arccos(cos_result))
angle
```

### 内积正、负、零

两个向量内积结果为正，说明向量夹角在 0° 到 90° (包括 0°，不包括 90°)。
内积结果为0，说明两个向量垂直。
内积结果为负，向量夹角在 90° 到 180°(包括 180°，不包括 90°)。

## 22.5 二维到三维

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def draw_vector(ax, vector, color, label):
    """
    绘制3D向量并添加标签
    参数:
    - ax: 3D子图对象
    - vector: 向量数组，形如 [x, y, z]
    - color: 向量颜色
    - label: 向量标签
    """
    ax.quiver(0, 0, 0, vector[0], vector[1], vector[2],
             color=color, arrow_length_ratio=0.1, normalize=False)
    label_text = f"{label}, ({vector[0]},{vector[1]},{vector[2]})"
    ax.text(vector[0], vector[1], vector[2], label_text, color=color, verticalalignment="center")
```

```python
c = np.array([4, 3, 5])  
a = np.array([4, 3, 0])  
i = np.array([1, 0, 0])  
j = np.array([0, 1, 0])  
k = np.array([0, 0, 1])  
```

```python
fig = plt.figure()
ax = plt.axes(projection="3d")

draw_vector(ax, a, "blue", "a")
draw_vector(ax, c, "green", "c")
draw_vector(ax, i, "red", "i")
draw_vector(ax, j, "purple", "j")
draw_vector(ax, k, "orange", "k")

ax.set_proj_type("ortho")

# 设置坐标范围
ax.set_xlim(0, 5)
ax.set_ylim(0, 5)
ax.set_zlim(0, 5)

ax.set_xlabel(r"$\it{x}$")
ax.set_ylabel(r"$\it{y}$")
ax.set_xlabel(r"$\it{z}$")

ax.view_init(azim=60, elev=20)

ax.xaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})
ax.yaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})
ax.zaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})

plt.tight_layout()
plt.show()
```

## 22.6 投影：影子的长度

投影分为两种——<font color="blue">标量投影</font>(scalar projection)和<font color="blue">向量投影</font>(vector projection)。

### 标量投影

向量标量投影结果为标量。
利用向量内积的计算原理，向量 $\boldsymbol{b}$ 在向量 $\boldsymbol{a}$ 方向上投影得到的线段长度就是标量投影:
$$
\|\boldsymbol{b}\|\cos\theta=
\|\boldsymbol{b}\|
\frac{\boldsymbol{a}\cdot\boldsymbol{b}}
{\|\boldsymbol{a}\|\|\boldsymbol{b}\|}=
\frac{\boldsymbol{a}\cdot\boldsymbol{b}}
{\|\boldsymbol{a}\|}
$$



### 向量投影

向量投影则是在标量基础上，加上方向。也就是说 $\boldsymbol{b}$ 在 $\boldsymbol{a}$ 方向上标量投影，再乘 $\boldsymbol{a}$ 的单位向量:
$$
\operatorname{proj}_{\boldsymbol{a}}\boldsymbol{b}=
\|\boldsymbol{b}\|\cos\theta
\frac{\boldsymbol{a}}{\|\boldsymbol{a}\|}=
\frac{\boldsymbol{a}\cdot\boldsymbol{b}}{\|\boldsymbol{a}\|}
\frac{\boldsymbol{a}}{\|\boldsymbol{a}\|}
$$
