### 乘法

乘法符号× 用于数字相乘，一般不用在两个变量相乘。在线性代数中，×则表示<font color="blue">叉乘</font>(cross product 或 vector product), 完全另外一回事

$a · b$ 表示 $a$ 和 $b$ 两个向量的<font color="blue">标量积</font>(scalar product)。 

```python
num1 = 2
num2 = 3

# multiplied two numbers
prod = num1 * num2

# display the computation
print(f"The product of {num1} and {num2} is {prod}")
```

### 阶乘

某个正整数的<font color="blue">阶乘</font>(factorcial)是所有小于及等于该数的正整数的积。  
<font color="red">0 的阶乘为 0! = 1</font>

```python
# define a function to calculate factorial
num = int(input("Enter an integer:"))

factorial = 1

# check if the number is negative, positive or zero
if num < 0:
    print("Factorial does not exist for negative numbers")
elif num == 0:
    print(f"The factorial of 0 is {factorial}")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"The factorial of {num} is {factorial}")
```
