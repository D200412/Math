## 4.2 集合：确定的一堆东西

<font color="blue">集合</font>(set)是由一个或多个确定的<font color="blue">元素</font>(member 或 element)所构成的整体。集合可以分为：<font color="blue">有限元素集合</font>(finite set)、<font color="blue">无限元素集合</font>(infinite set)和<font color="blue">空集</font>(empty set 或 null set)。

### 集合与元素

集合与元素的关系有两种：<font color="blue">属于</font>(belongto) $\in$，<font color="blue">不属于</font>(not belongto) $\notin$。

### 集合和集合

如果集合 $A$ 中的每一个元素也都是集合 $B$ 中的元素，那么 $A$ 是 $B$ 的<font color="blue">子集</font>(subset)，记作$A \subseteq B$。而 $B$ 是 $A$ 的<font color="blue">母集</font>，也称<font color="blue">超集</font>(superset)。

如果同时满足 $A \subseteq B$ 和 $A≠B$，则称 $A$ 是 $B$ 的<font color="blue">真子集</font>(Aisaproper subset of B)，记作 $A \subset B$。

给定 $A$ 和 $B$ 两个集合，由所有属于 $A$ 且属于 $B$ 的元素所组成的集合，叫做 $A$ 与 $B$ 的<font color="blue">交集</font>(intersection), 记作 $A \cap B$。

$A$ 和 $B$ 所有的元素合并组成的集合，叫做 $A$ 和 $B$ 的<font color="blue">并集</font>(union)，记作 $A \cup B$。

<font color="blue">补集</font>(complement)一般指<font color="blue">绝对补集</font><(absolute complement)。设 $\Omega$ 是一个集合，$A$ 是 $\Omega$ 的一个子集，由 $\Omega$ 中所有不属于 $A$ 的元素组成的集合，叫做子集 $A$ 在 $\Omega$ 中的绝对补集。

$A$ 中 $B$ 的<font color="blue">相对补集</font>(relative complement)，是所有属于$A$ 但不属于 $B$ 的元素组成的集合，记做 $A \setminus  B$ 或 $A-B$ (set difference of $A$ and $B$), 可以读作“ $B$ 在 $A$ 中的相对补集(the relative complement of $B$ with respect to set $A$)"。

### 文氏图

### 掷骰子

```python
# set A: odd
A  = set([1, 3, 5])

# set B: even
B = set([2, 4, 6])

# set C: less than 4
C = set([1, 2, 3])

# A union B
A_union_B = A.union(B) # A|B, A or B

# A instersects (meets) B
A_meet_B = A.intersection(B) # A&B, A and B

# A minus B
A_minus_B = A.difference(B) # A - B, set difference

# A union C
A_union_C = A.union(C) # A|C, A or C

# A intersects (meets) C
A_meet_C = A.intersection(C) # A&C, A and C

# A minus C
A_minus_C = A.difference(C) # A - C, set difference

# C minus A
C_minus_A = C.difference(A) # C - A, set difference
```

## 4.3 从代数式到函数

<font color="blue">算数</font>(arithmetic)基于<font color="blue">已知量</font>(known values)；而<font color="blue">代数</font>(algebra)基于<font color="blue">未知量</font>(unknown values)。也称作<font color="blue">变量</font>(variables)。

### 多项式

最常见的代数式是<font color="blue">多项式</font>(polynomial)。



其中，$x$ 为<font color="blue">变量</font>(variable), $n$ 为<font color="blue">多项式次数</font>(degree of a polynomial), $a_0、a_1 ... a_n$ 为<font color="blue">系数</font>(coefficient).

```python
from sympy.abc import x, y

expr_x = x ** 3 + 2 * x ** 2 - x - 2

# subs() 把 x 替换成 1 
print(expr_x.subs(x, 1))
```

```python
from sympy import cos

expr_cos_y = expr_x.subs(x, cos(y))

print(expr_cos_y)
```

```python
from sympy import symbols

x, y = symbols("x,y")
expr_1 = x + y
print(expr_1)

x1, x2 = symbols("x1,x2")
expr_2 = x1 + x2
print(expr_2)
```

```python
from sympy import *
x, y, z = symbols("x y z")

# simplify mathemarical expressions
expr_1 = sin(x) ** 2 - cos(x) ** 2
# sympy.simplify() 函数可以简化代数式
print(simplify(expr_1))

# expand polynomial expressions
expr_2 = (x + 1) ** 3
# sympy.expand() 可以用来展开代数式
print(expand(expr_2))

# take a polynomial and factors. it into irreducible factors
expr_3 = x ** 3 + 2 * x ** 2 - x - 2 
# sympy.factor() 函数可以对代数式因式分解
print(factor(expr_3))

# collect common powers of a term in an expression
# sympy.collect() 函数用来合并同类项。
expr_4 = collect(expr_3 - x ** 2 - 2 * x, x) 
print(expr_4)
```

### 函数

给定一个集合 $X$，对 $X$ 中元素 $x$ 施加映射法则 $f$，记作函数 $fx)$，得到的结果$y=f(x)$ 属于集合 $Y$。集合 $X$ 称作<font color="blue">定义域</font>(domain)，$Y$ 称为<font color="blue">值域</font>(codomain)；$x$ 称作<font color="blue">自变量</font>(an argument of a function或an independent variable)，$y$ 称作<font color="blue">因变量</font>(dependent variable)

函数的自变量为两个或两个以上时，叫做<font color="blue">多元函数</font>(multivaraite function)。

```python
from sympy.abc import x, y, z 
from sympy.utilities.lambdify import lambdify

expr = x ** 3 +  2 * x ** 2 - x - 2
f_x = lambdify(x, expr)

print(f_x(1))
```

## 4.4 杨辉三角：代数和几何的完美合体

<font color="blue">火柴梗图</font>(stem plot)方便可视化杨辉三角每行单项式系数的规律

```python
# 绘制火柴梗图
from sympy.abc import x, a
from sympy import Poly
import matplotlib.pyplot as plt
import numpy as np

for n in [4, 8, 12, 5, 9, 13, 36]:
    expr = (x + 1) ** n # 定义多项式表达式

    expr_expand = expr.expand()
    expr_expand = Poly(expr_expand)

    # 获取多项式的系数
    poly_coeffs = expr_expand.coeffs()

    print(poly_coeffs)

    degrees = np.linspace(n, 0, n+1)

    fig, ax = plt.subplots()

    plt.stem(degrees, poly_coeffs)
    plt.xlim(0, n)
    plt.xticks(np.arange(0, n+1))
    y_max = max(poly_coeffs)
    y_max = float(y_max)
    plt.ylim(0, y_max)

    ax.spines["right"].set_visible(False)
    ax.spines["top"].set_visible(False)
    # 反转 x 轴
    ax.invert_xaxis()
    plt.xlabel("Degree")
    plt.ylabel("Coefficient")
    plt.show()
```

## 4.5 排列组合让系数更具意义

$n$ 个不同元素中取出 $m$ 个元素的所有组合个数叫做<font color="blue">组合</font>(combination)数，常记做 $C^m_n$


```python
import itertools

letters = "ABC"

# find all combinations containing 2 letters

cmb = itertools.combinations(letters, 2)

for val in cmb:
    print(val)
```

### 组合数表达杨辉三角

### 排列数

从 $n$ 个不同元素中，先后取$m$ ( $m≤n$ )个元素排成一列，叫做从 $n$ 个元素中取出 $m$个元素的一个<font color="blue">排列</font>(permutation)。排列数常被记做 $P^m_n$

```python
import itertools

letters = "ABC"

# Arranging 2 elemrnts out of 3

per = itertools.permutations(letters, 2)

for val in per:
    print(val)
```

### 组合数和排列数

排列和组合的关系



$m = n$ 时，所有的排列情况叫全排列。



```python
import itertools

letters = "ABC"

# Aranging all 3 elements

per = itertools.permutations(letters)

for val in per:
    print(val)
```
