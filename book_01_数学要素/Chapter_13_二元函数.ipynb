# 13 二元函数(Bivariate Functions)

## 13.1 二元一次函数：平面

二元一次函数是一元一次函数的扩展，一般式如下:
$$
y=f(x_1,x_2)=w_1x_1+w_2x_2+b
$$

用矩阵乘法，可以写成: 
$$
y=f(x_1,x_2)=\mathbf{w}^\mathrm{T}\mathbf{x}+b
$$
其中
$$
\mathbf{w} = \begin{bmatrix}w_1\\w_2\end{bmatrix},
\mathbf{x} = \begin{bmatrix}x_1 \\ x_2\end{bmatrix}
$$

### 等高线平行纵轴

当 $w_1=-1$，$w_2=0$，$b=0$ 时，$f(x_1,x_2)$ 平面仅仅受到 $x_1$ 影响。解析式:
$$
y=f(x_1,x_2)=-x_1=\begin{bmatrix} -1 \\ 0 \end{bmatrix}^\mathrm{T}\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
$$

### 等高线平行横轴

当 $w_1=0$，$w_2=1$，$b=0$ 时，$f(x_1,x_2)$ 平面仅仅受到 $x_2$ 影响。解析式:
$$
y=f(x_1,x_2)=x_2=\begin{bmatrix} 0 \\ 1 \end{bmatrix}^\mathrm{T}\begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
$$

### 平面叠加

### 超平面

一次函数中变量数量继续增多时，将获得<font color="blue">超平面</font>(byperplane)，对应的解析式如下:
$$
y=f(x_1,x_2,…,x_D)=w_1x_1+w_2x_2+…+w_Dx_D+b
$$

```python
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
```

```python
## 生成正方形网格
def mesh_square(x1_0, x2_0, r, num):
    rr = np.linspace(-r, r, num)
    xx1, xx2 = np.meshgrid(rr, rr)

    xx1 = xx1 + x1_0
    xx2 = xx2 + x2_0

    return xx1, xx2
```

```python
# 绘制三维曲面图
def plot_suf(xx1, xx2, ff, caption):
    norm_plt = plt.Normalize(ff.min(), ff.max())
    colors = cm.coolwarm(norm_plt(ff))

    fig = plt.figure()
    ax = plt.axes(projection="3d")
    surf = ax.plot_surface(xx1, xx2, ff, facecolors=colors, shade=False) # 绘制曲面
    surf.set_facecolor((0, 0, 0, 0))

    plt.tight_layout()
    ax.set_xlabel(r"$\it{x_1}$")
    ax.set_ylabel(r"$\it{x_2}$")
    ax.set_zlabel(r"$\it{f}$($\it{x_1}$,$\it{x_2}$)")
    ax.set_title(caption)
    ax.set_proj_type("ortho")
    ax.set_xlim(xx1.min(), xx1.max())
    ax.set_ylim(xx2.min(), xx2.max())
    ax.xaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})
    ax.yaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})
    ax.zaxis._axinfo["grid"].update({"linewidth":0.25, "linestyle":":"})
```

```python
# 绘制等高线填充图
def plot_contourf(xx1, xx2, ff, caption):
    fig, ax = plt.subplots()
    cntr2 = ax.contourf(xx1, xx2, ff, levels=15, cmap="RdBu_r")

    fig.colorbar(cntr2, ax=ax)
    ax.set_xlabel(r"$\it{x_1}$")
    ax.set_ylabel(r"$\it{x_2}$")
    ax.set_title(caption)
    ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
```

```python
x1_0 = 0
x2_0 = 0
r = 2
num = 30
xx1, xx2 = mesh_square(x1_0, x2_0, r, num)
```

```python
# f(x1, x2) = -x1
ff = -xx1
caption= r"$\it{f}=-\it{x_1}$"

plot_suf(xx1, xx2, ff, caption)
plot_contourf(xx1, xx2, ff, caption)
```

```python
# f(x1, x2) = x2
ff = xx2
caption = r"$\it{f} = \it{xx_2}$"

plot_suf(xx1, xx2, ff, caption)
plot_contourf(xx1, xx2, ff, caption)
```

```python
# f(x1,x2) = x1 + x2
ff = xx1 + xx2
caption = r"$\it{f} = \it{x_1} + \it{x_2}$"

plot_suf(xx1, xx2, ff, caption)
plot_contourf(xx1, xx2, ff, caption)
```

```python
# f(x1,x2)  = -x1 + x2
ff = -xx1 + xx2
caption = r"$\it{f} = -\it{x_1} + \it{xx_2}$"

plot_suf(xx1, xx2, ff, caption)
plot_contourf(xx1, xx2, ff, caption)
```

## 13.2 正圆抛物面：等高线为正圆

<font color="blue">正圆抛物面</font>(circular paraboloid)的是<font color="blue">抛物面</font>(paraboloid)的一种特殊形式，它的等高线为正圆。正圆抛物面的最简单的形式为:
$$
y=f(x_1,x_2)=a(x_1^2+x_2^2)
$$
可以写成如下矩阵运算形式:
$$
y = f(x_1,x_2)
= \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
  \begin{bmatrix} a & 0 \\ 0 & a \end{bmatrix}
  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
= a \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
    \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
= a\mathbf{x}^{\mathrm{T}}\mathbf{x}
= a\lVert \mathbf{x} \rVert^{2}
$$

### 向量的模

$\|\mathbf{x}\|$ 为向量 $\mathbf{x}$ 的<font color="blue">模</font>(norm)，就是向量的长度，定义为：
$$
\|\mathbf{x}\|=\sqrt{x_1^2+x^2_2}\|
$$

### 开口朝上

$$
y = f(x_1 = 1, x_2) = 1+x_2^2
$$


### 开口朝下

$$
y = f(x_1,x_2)
= -x_1^2-x_2^2
= \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
  \begin{bmatrix} -1 & 0 \\ 0 & -1 \end{bmatrix}
  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
= -\mathbf{x}^{\mathrm{T}}\mathbf{x}
$$


### 平移

在三维直角坐标系中，二元函数变量 $(x_1,x_2)$ 平移 $(c_1,c_2)$ 得到:
$$
y=f(x_1,x_2)=-(x_1-c_1)^2-(x_2-c_2)^2=-(\mathbf{x}-\mathbf{c})^{\mathrm{T}}(\mathbf{x}-\mathbf{c})=-\|\mathbf{x}-\mathbf{c}\|^2
$$
其中，$\mathbf{c}=[c_1,c_2]^\mathrm{T}$



## 13.3 椭圆抛物面：等高线为椭圆

### 开口朝上

开口朝上<font color="blue">椭圆抛物面</font>(elliptic paraboloid)的一般形式为:
$$
y = f(x_1,x_2)
= \frac{x_1^2}{a^2} + \frac{x_2^2}{b^2}
= \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
  \begin{bmatrix}
    \frac{1}{a^2} & 0 \\
    0 & \frac{1}{b^2}
  \end{bmatrix}
  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
$$
其中，$a$ 和 $b$ 都不为 0。特别地，当 $a^2=b^2$ 时，椭圆抛物面便是正圆抛物面。


### 开口朝下

$$
y = f(x_1,x_2)
= -\frac{x_1^2}{a^2} - \frac{x_2^2}{b^2}
= \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
  \begin{bmatrix}
    -\frac{1}{a^2} & 0 \\
    0 & -\frac{1}{b^2}
  \end{bmatrix}
  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
$$


### 旋转

<font color="red">和矩阵分解中的特征值分解有关系，和正定性有关系，和协方差矩阵有关系，和线性变换(仿射变换)有关系</font>

<font color="green">在多元线性回归中，为了简化模型复杂度，可以引入正则项(regularizer)。正则项的目的是"收缩"，即让某些估计参数变小，甚至为 0。</font>

## 13.4 双曲抛物面：马鞍面

<font color="blue">双曲抛物面</font>(hyperbolic paraboloid)，也叫<font color="blue">马鞍面</font>(saddle surface)，其形状酷似马鞍而得名。双曲抛物面的一般形式为
$$
y = f(x_1,x_2)
= \frac{x_1^2}{a^2} - \frac{x_2^2}{b^2}
= \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}^{\mathrm{T}}
  \begin{bmatrix}
    \frac{1}{a^2} & 0 \\
    0 & -\frac{1}{b^2}
  \end{bmatrix}
  \begin{bmatrix} x_1 \\ x_2 \end{bmatrix}
$$

曲面等高线为一系列双曲线。而曲面中心点，也称作<font color="blue">鞍点</font>(saddle point)，即不是曲面的最大值点也不是最小值点。

## 13.6 锥面：正圆抛物面开方

### 开口朝上

开口朝上正圆抛物面解析式开平方取正，便得到锥面。<font color="blue">锥面</font>(cone surface)开口朝上，对应解析式如下:
$$
y=f(x_1,x_2)=\sqrt{x_1^2+x_2^2}=\sqrt{\mathbf{x}^{mathrm{T}}\mathbf{x}}=\|\mathbf{x}\|
$$
锥面的等高线为一系列同心圆。

### 开口朝下

解析式加个负号便得到开口向下锥面，解析式如下:
$$
y=f(x_1,x_2)=-\sqrt{x_1^2+x_2^2}
$$


### 对顶圆锥

中轴保持在一条直线上，将开口朝上和开口朝下两个圆锥面在顶点处拼接在一起便获得<font color="blue">对顶圆锥</font>(double cone 或 vertically opposite circular cone)。



<font color="green">如果两个圆锥面底对底拼接在一起便获得了双锥体</font>

### 13.7 绝对值函数：和超椭圆有关

### 平面对折

$x_1+x_2$ 取绝对值，具体解析式如下:
$$
y =f(x_1, x_2) = |x_1 + x_2|
$$


### 旋转正方形

$x_1$ 和 $x_2$ 分别取绝对值再求和，解析式如下:
$$
y=f(x_1,x_2)=|x_1|+|x_2|
$$


<font color="red">和 $L^p$ 范数有关系，和距离度量有关系</font>

### 正方形

$x_1$ 和 $x_2$ 分别取绝对值，比大小后、取两者中最大值，如下:
$$
y=f(x_1,x_2)=max(|x_1|,|x_2|)
$$


## 13.8 逻辑函数：从一元到二元 

二元逻辑函数对应的一般解析式如下:
$$
y=f(x_1, x_2)=\frac{1}{1+\exp(-(w_1x_1+w_2x_2+b))}
$$
写成矩阵运算形式:
$$
y=f(x)=\frac{1}{1+\exp(-(\mathbf{w}^{\mathrm{T}}\mathbf{x}+b))}
$$
当 $x_1+x_2$ 趋近正无穷时，趋近1，却无法达到1。当 $x_1+x_2$ 趋向于负无穷时，趋近0，却无法达到0。



当 $w_1$ 和 $w_2$ 增大后，坡面变得陡峭。


### 二元 tanh() 函数

二元 $\tanh()$ 函数形式如下:
$$
y=f(x_1,x_2)=\tanh(\gamma(w_1x_1+w_2x_2)+r)
$$
写成矩阵运算形式:
$$
y=f(x)=\tanh(\gamma \mathbf{w}^\mathrm{T}\mathbf{x}+r)
$$
当 $\gamma$ 增大时，曲面也变得陡峭。


## 13.9 高斯函数：机器学习的多面手

二元高斯函数的一般形式为:
$$
y=f(x_1,x_2)=\exp(-\gamma((x_1-c_1)^2+(x_2-c_2)^2))
$$

可以发现随着 $\gamma$ 增大，曲面变得更尖，更陡峭。


<font color="red">与kernel trick有关系，与二元的高斯分布有关，与协方差矩阵有关</font>
