## 1.2 数字分类：从复数到自然数
复数(complex number)包括实数(real number)和虚数(imaginary number)。 
$$
a + bi
$$
其中 $a$ 和 $b$ 是实数。$i$ 是虚数单位(imaginary unit)。  

实数包括有理数(rational numbers)和无理数(irrational numbers)。  

有理数可以表示为两个整数的商(quotient of two integers)
$$
\frac{a}{b} 
$$

```python
# Import math library
import math

# Print the value of pi
print("pi =", math.pi)

# Print the value of e
print("e =", math.e)

# Print the value of square root of 2
print("sqrt(2) =" , math.sqrt(2))
```

### 判断数字奇偶性

```python
# A number is even if division by 2 gives a remainder of 0.
# If the remainder is 1, it is an odd number
num = float(input("Enter a number:"))

if num.is_integer():

    if (num % 2) == 0:
        print(f"{int(num)} is a even.")
    else:
        print(f"{int(num)} is a odd.")
else:
    print(f"{num} is not a integer")
```

## 1.4 向量：数字排成行、列


若干数字排成一行或一列，并且用中括号括起来，得到的数组叫做<font color="blue">向量</font>(vectoer)  
排成一行的叫做<font color="blue">行向量</font>(row vector)，排成一列的叫做<font color="blue">列向量</font>(column vector)  
行向量是表格的一行数字，列向量是表格的一列数字  
  
用 `numpy.array([[1, 2, 3]])` 定义行向量  
用 `numpy.array([[1], [2], [3]])` 定义列向量

```python
import numpy as np

# row vector
a = np.array([[1, 2, 3]])
print(a)

# column vecctor
b = np.array([[1], [2], [3]])
print(b)
```

### 转置

行向量<font color="blue">转置</font>(transpose)得到列向量；列向量转置得到行向量  
  

```python
import numpy as np

# row vector transposed to a column vector
a_row = np.array([[1, 2, 3]])

b = a_row.T

b_col = np.array([[1], [2], [3]])

a = b_col.T

print(a_row)
print(b)
print(b_col)
print(a)
```
