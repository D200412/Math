## 9.2 离心率：联系不同类型圆锥曲线

不同类型圆锥曲线可以通过同<font color="blue">离心率</font>(eccentricity) $e$ 联系起来:
$$
x^2_2=2px_1+(e^2-1)x_1^2,e≥0
$$
正圆的离心率 $e=0$，椭圆的离心率 $0<e<1$，抛物线离心率 $e=1$，双曲线离心率 $e>1$。

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as patches

x = np.linspace(-4, 4, num=201)
y = np.linspace(-4, 4, num=201)
m = 1
n = 1.5

xx, yy = np.meshgrid(x, y)
e_array = np.linspace(0, 3, num=31)

fig, ax = plt.subplots(figsize=(8, 8))

colors = plt.cm.RdYlBu(np.linspace(0, 1, len(e_array)))

for i in range(0, len(e_array)):
    e = e_array[i]
    ellipse = yy ** 2 - (e ** 2 - 1) * xx ** 2 - 2 * xx
    color_code = colors[i, :].tolist()
    plt.contour(xx, yy, ellipse, levels=[0], colors=[color_code])

plt.axvline(x=0, color="k", linestyle="-")
plt.axhline(y=0, color="k", linestyle="-")
ax.set_xticks([])
ax.set_yticks([])
ax.set_xlim([-4, 4])
ax.set_ylim([-4, 4])

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)
```

上式可以看做是椭圆和双曲线的“叠加”。$x_1x_2=1$ 实际上是一个旋转双曲线。参数 $\rho$ 可以视作调节双曲线“影响力"的参数，$\rho$ 越大双曲线的影响越强。

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.patches as parches

a = 2
b = 1

x = np.linspace(-4, 4, num=201)
y = np.linspace(-4, 4, num=201)
xx, yy = np.meshgrid(x, y)

k_array = np.linspace(0, -2, num=21)

fig, ax = plt.subplots(figsize=(8, 8))

colors = plt.cm.RdYlBu(np.linspace(0, 1, len(k_array)))

for i in range(0, len(k_array)):
    k = k_array[i]
    ellipse = (xx / a) ** 2 + (yy / b) ** 2 - 2 * k * (xx / a) * (yy / b)
    color_code = colors[i,:].tolist()
    plt.contour(xx, yy, ellipse, levels=[1], colors=[color_code])

plt.axvline(x=0, color="k", linestyle="-")
plt.axhline(y=0, color="k", linestyle="-")
plt.xticks([])
plt.yticks([])
plt.xlim(-4, 4)
plt.ylim(-4, 4)

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)
```

```python
import matplotlib.pyplot as plt
import numpy as np

a = 1
b = 1

p = [0.5, 1, 2, 3]
q = p

pp, qq = np.meshgrid(p, q)
# 将网格展平成一维数组
pp = pp.flatten()
qq = qq.flatten()

x1 = np.linspace(-2, 2, num=101)
x2 = x1

xx1, xx2 = np.meshgrid(x1, x2)

fig, axes = plt.subplots(ncols=4, nrows=4, figsize=(12, 12))

for p, q, ax in zip(pp, qq, axes.flat):
    if np.isinf(p):  # 当 p 为无穷大时，计算 Lp 空间的等值线
        zz = np.maximum(np.abs(xx1 / a), np.abs(xx2 / b))
    else:
        zz = (np.abs(xx1 / a) ** p + np.abs(xx2 / b) ** q) **(1 / q)

    ax.contourf(xx1, xx2, zz, 20, cmap="RdYlBu_r")
    ax.contour(xx1, xx2, zz, [1], colors="k", linewidth=2)

    ax.axhline(y=0, color="k", linewidth=0.25)
    ax.axvline(x=0, color="k", linewidth=0.25)
    ax.set_xlim(-2, 2)
    ax.set_ylim(-2, 2)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["bottom"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_title("p = " + str(p) + ", q = " + str(q))
    ax.set_aspect("equal", adjustable="box")
    
plt.show()
```

## 9.7 圆锥曲线一般式

圆锥曲线的一般式形如下:
$$
Ax^2_1+Bx_1x_2+Cx^2_2+Dx_1+Ex_2+F=0
$$

满足下列条件，圆锥曲线为正圆:
$$
Ax^2_1+Cx^2_2+Dx_1+Ex_2+F=0, \; A=C
$$

满足下列条件，圆锥曲线为正椭圆，即没有旋转:
$$
Ax^2_1+Cx^2_2+Dx_1+Ex_2+F=0, \; A≠C, \; AC>0
$$

满足下列条件，圆锥曲线为正双曲线:
$$
Ax^2_1+Cx^2_2+Dx_1+Ex_2+F=0, \; AC<0
$$

满足下列任一等式，圆锥曲线为正抛物线:
$$
\left\{
\begin{matrix}
 Ax^2_2+Dx_1+Ex_2+F=0
 \\
Cx^2_2+Dx_1+Ex_2+F=0
\end{matrix}
\right.
$$

- $B^2-4AC<0$ 时，圆锥曲线为椭圆
- $B^2-4AC=0$ 时，圆锥曲线为抛物线
- $B^2-44C>0$ 时，圆锥曲线为双曲线。

### 矩阵运算

矩阵运算式：



```python
import matplotlib.pyplot as plt
import numpy as np

# 圆锥曲线的一般形式
# Ax1^2 + Bx1x2 + Cx2^2 + Dx1 + Ex2 + F = 0

def conic_plot(A, B, C, D, E, F, title):
    x1 = np.linspace(-5, 5, 500)
    x2 = np.linspace(-5, 5, 500)
    x1, x2 = np.meshgrid(x1, x2)

    z = A * x1 ** 2 + B * x1 * x2 + C * x2 ** 2 +  D * x1 + E * x2 + F

    fig, ax = plt.subplots(figsize=(6, 6))
    ax.contour(x1, x2, z, levels=[0], colors="blue")
    plt.title(title)
    plt.xlabel("x1")
    plt.ylabel("x2")
    ax.set_aspect("equal", "box")
    ax.set_xlim(-5, 5)
    ax.set_ylim(-5, 5)
    ax.set_xticks(np.arange(-5, 6, 1))
    ax.set_yticks(np.arange(-5, 6, 1))
    plt.axvline(x=0, color="black", linewidth=0.5)
    plt.axhline(y=0, color="black", linewidth=0.5)
    plt.grid(True, linestyle="-", color="0.8", alpha=0.7)
    plt.savefig("09_" + title + ".svg")
    plt.show()
```

```python
# 示例 1: 正圆 A = 1, B = 0, C = 1, D = 0, E = 0, F = -9
conic_plot(1, 0, 1, 0, 0, -9, "Circle")
```

```python
# 示例 2: 椭圆 A = 4, B = 0, C = 9, D = 0, E = 0, F = -36
conic_plot(4, 0, 9, 0, 0, -36, "Ellipse")
```

```python
# 示例3: 旋转椭圆 A = 5, B = 4, C = 5, D = 0, E = 0, F = -50
conic_plot(5, 4, 5, 0, 0, -50, "Rotated Ellipse")
```

```python
# 示例 4: 抛物线 A = 0, B = 0, C = 1, D = -1, E = 0, F = -4
conic_plot(0, 0, 1, -1, 0, -4, "Parabola")
```
