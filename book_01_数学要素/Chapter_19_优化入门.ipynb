# 19 优化入门(Fundamebtals of Optimization)

## 19.1 优化问题：寻找山峰、山谷

简单地说，<font color="blue">优化问题</font>(optimization problem)是在给定约束条件下改变<font color="blue">变量</font>(variable或 optimization variable)，用某种数学方法，寻找特定目标的<font colot="blue">最优解</font>(optimized solution或 optimal solution或 optimum)。



山谷对应<font color="blue">极小值</font>(minima 或 local minima 或 relative minima)，山峰对应<font color="blue">极大值</font>(maxima 或 local maxima 或 relative maxima).

### 极值

从一元函数角度讲，函数 $f(x)$ 在 $x=a$ 点的某个邻域内有定义，对于 $a$ 的去心邻域内任一 $x$ 满足:
$$
f(a) < f(x)
$$
就称 $f(a)$ 是函数的极小值。也可以说，$f(x)$ 在 $a$ 点处取得极小值，$x=a$ 是函数 $f(x)$ 的极值点。

相反，如果 $a$ 的去心邻域内任一 $x$ 满足:
$$
f(a)>f(x)
$$
$f(a)$ 是函数的极大值，即 $f(x)$ 在 $a$ 点处取得极大值，同样 $x=a$ 也是函数 $f(x)$ 的极值点。

<font color="blue">极值</font>(extrema 或 local extrema)是<font color="blue">极大值</font>和<font color="blue">极小值</font>的统称。

### 最值

如果某个极值是整个指定搜索区域内的极大值或极小值，这个极值又被称作是<font color="blue">最大值</font>(maximum 或 global maximum)或者<font colot="blue">最小值</font>(minimum 或 global minimum).

最大值和最小值统称最值(global extrema)。

从一元函数角度讲，函数 $f(x)$ 在整个搜索域内有定义，对于搜索域内任一 $x$ 满足:
$$
f(a)<f(x)
$$
就称 $f(a)$ 是函数的最小值，$x=a$ 是函数 $f(x)$ 的全局最优解。

## 19.2 构造优化问题

### 最小化优化问题

最小化优化问题可以写成:
$$
\underset{x}{\arg\min}\; f(x)
$$

$\arg\min$ 的含义是 argument of the minima；$x$ 是自变量，多变量一般写成列向量；$f(x)$ 为<font color="blue">目标函数</font>(objective function)，它可以是个函数解析式(比如 $f(x)=x_2$ )，也可以是个无法用解析式表达的模型。

### 优化变量

$x$ 为优化变量，也叫决策变量。优化变量可以是一个未知量 $x$，优化变量也可以是多个未知量构成的列向量，比如 $x=[x_1,x_2,...x_D]^{\mathrm{T}}$。

如果优化问题的变量只有一个，这类优化叫做<font color="blue">单变量优化</font>(single-variableoptimization)；如果优化问题有多个变量，优化问题叫<font color="blue">多变量优化</font>(multi-variable optimization)。

当只有一个优化变量时候，目标函数可以写成一元函数 $f(x)$，$f(x)$ 和变量 $x$ 的关系可以在平面
上表达。

有两个优化变量的情况，比如 $x_1$ 和 $x_2$，目标函数为二元函数 $f(x_1,x_2)$ 时，$f(x_1,x_2)$、$x_1$ 和 $x_2$ 的关系可以利用三维等高线表达。

平面等高线也可以展示两个优化变量优化问题。

当优化变量数量不断增多，优化变量要写成列向量 $\mathbf{x}=[x_1,x_2,...,x_{D}]^{\mathrm{T}}$，$f(x)$ 在多维空间中形成一个<font color="blue">超曲面</font>(hypersurface)。

此外，优化目标可以有一个或多个。具有不止一个目标函数优化问题叫做<font color="blue">多目标优化</font>(multiobjective optimization)。

### 最大化优化问题

最大化优化问题可以写成:
$$
\underset{x}{\arg\max}\; f(x)
$$

实际上，标准优化问题一般都是最小化优化问题。而最大化问题目标函数改变符号(乘 -1 ),就转化为最小化问题:
$$
\underset{x}{\arg\max}\; f(x)
\quad\Longleftrightarrow\quad
\underset{x}{\arg\min}\; -f(x)
$$



## 19.3 约束条件：限定搜索区域

优化变量取值并非随心所欲，必须在一定范围之内。变量的取值范围叫做<font color="blue">定义域</font>(domain)，也叫做<font color="blue">搜索空间</font>(search space)、<font color="blue">选择集</font>(choice set)。

范围内的每一个点为一个<font color="blue">潜在解</font>(candidate solution或feasible solution)。

优化变量取值范围的条件被称作<font color="blue">约束条件</font>(constraints)。根据约束条件的有无，优化问题分为:
- <font color="blue">无约束优化问题</font>(unconstrained optimization)
- <font color="blue">受约束优化问题</font>(constrained optimization) 

### 五类约束

多数优化问题都是受约束优化问题，常见的约束条件分为以下几种:
- <font color="blue">上下界</font>(lower and upper bounds)，$l≤x≤u$
- <font color="blue">线性不等式</font>(linear inequalities)，$g(x)=Ax-b≤0$
- <font color="blue">线性等式</font>(linear equalities)，$h(x)= A_{eg}x - b_{eq}=0$
- <font color="blue">非线性不等式</font>(nonlinear inequalities)，$c(x)≤0$
- <font color="blue">非线性等式</font>(nonlinear equalities)，$c_{eg}(x)=0$

### 最小化问题

结合约束条件，构造完整最小化问题:
$$
\begin{aligned}
&\underset{\boldsymbol{x}}{\arg\min}\; f(\boldsymbol{x})\\
&\text{subject to:}\quad
\boldsymbol{l}\leq\boldsymbol{x}\leq\boldsymbol{u}\\
&\phantom{\text{subject to:}\quad}
A\boldsymbol{x}-\boldsymbol{b}\leq 0\\
&\phantom{\text{subject to:}\quad}
A_{\mathrm{eq}}\boldsymbol{x}-\boldsymbol{b}_{\mathrm{eq}}=0\\
&\phantom{\text{subject to:}\quad}
c(\boldsymbol{x})\leq 0\\
&\phantom{\text{subject to:}\quad}
c_{\mathrm{eq}}(\boldsymbol{x})=0
\end{aligned}
$$
其中，subject to代表“受限于”、“约束于”，常简写成 $\text{s.t.}$。

## 19.4 一元函数极值点判定

### 一阶导数

对于一元函数 $f(x)$，函数一阶导数为 0 的点叫<font color="blue">驻点</font>(stationarypoint)；而驻点可能是一元函数的极大值、极小值，或者是鞍点。

如果函数 $f(x)$ 在 $x=a$ 处可导，且在该点取得极值，则一阶导数 $f'(a)=0$。

### 二阶导数

函数 $f(x)$ 在 $x=a$ 处有二阶导数，且一阶导数 $f(a)=0$，即 $x=a$ 为驻点；如果二阶导数$f''(a)>0$ ，函数 $f(x)$在 $x=a$ 取得极小值；如果二阶导数 $f''(a)<0$，函数 $f(x)$ 在 $x=a$ 取得极大值。


### 一阶导数和二阶导数均为 0

如果函数驻点的一阶导数和二阶导数都为 0，可以用驻点左右一阶导数符号判定极值。

### 寻找极值点

总结来说，寻找极值时大家需要注意三类点：1)驻点(一阶导数为0点)，C 和 D 点；2)不可导点；3)搜索区域边界点。
<font color="red">导数不存在有又分三种情况：1)间断点，A和B点；2)尖点，E点；3)切线竖直，即斜率为无穷，F点</font>


### 编程求解优化问题

```python
import matplotlib.pyplot as plt
from scipy import optimize
import numpy as np
from sympy import lambdify, diff, exp
from sympy.abc import x 
```

```python
f_x = -2 * x * exp(-x ** 2)
f_x
```

```python
obj_f = lambdify(x, f_x)
```

```python
# 优化求解目标函数的最小值
result = optimize.minimize_scalar(obj_f) # 使用数值方法最小化目标函数

x_min = result.x # 提取最小值点 x 的位置

x_array = np.linspace(-2, 2 ,100)
y_array = obj_f(x_array)
```

```python
# 绘制目标函数图像及最小值点

fig, ax = plt.subplots()
plt.plot(x_array, y_array, color="b")
plt.plot(x_min, obj_f(x_min), color="r", marker="x", markersize=12)

plt.xlabel("x")
plt.ylabel("f(x)")
plt.xticks(np.linspace(-2, 2, 5))
plt.yticks(np.linspace(-2, 2, 5))
plt.axis("scaled")
ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.spines["top"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
```

```python
f_x_1_diff = diff(f_x, x) # 计算 f(x) 的一阶导数
f_x_1_diff
```

```python
f_x_1_diff_fcn= lambdify(x, f_x_1_diff)
f_x_1_diff_array = f_x_1_diff_fcn(x_array)

fig, ax = plt.subplots()
plt.plot(x_array, f_x_1_diff_array, color="b")
plt.plot(x_min, f_x_1_diff_fcn(x_min), color="r", marker="x", markersize=12)

plt.xlabel("x")
plt.ylabel("f'(x)")
plt.xticks(np.linspace(-2, 2, 5))
plt.yticks(np.linspace(-2, 2, 5))
plt.axis("scaled")
ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)
ax.spines["top"].set_visible(False)
ax.spines["bottom"].set_visible(False)
ax.spines["left"].set_visible(False)
ax.spines["right"].set_visible(False)
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
```

## 19.5 二元函数的极值点判定

### 一阶偏导

二元函数 $y=f(x_1,x_2)$ 在点 $(a,b)$ 存在分别对 $x_1$ 和 $x_2$ 存在偏导，且在 $f(a,b)$ 处有极值，则，
$$
f_{x_1}(a,b)=0, f_{x_2}(a,b)=0
$$

对于二元函数极值的判定，一阶偏导数 $f_{x_1}(x_1,x_2)=0$ 和 $f_{x_2}(x_1,x_2)=0$ 同时成立的点$(x_1,x_2)$ 为二元函数 $f(x_1,x_2)$ 的驻点。驻点可以是极小值、极大值或鞍点。


### 二阶偏导

如果 $f(xi,x2)$ 在 $(a,b)$ 邻域内连续，且函数的一阶偏导及二阶偏导连续，令，
$$
A=f_{x_1x_1}(a,b), B=f_{x_1x_2}(a,b), C=f_{x_2x_2}(a,b)
$$

$f(x_1,x_2)$ 在 $(a,b)$ 一阶偏导为 0，$f_{x_1}i(a,b)=0$，$f_{x_2}(a,b)=0$，$f(a,b)$是否为极值点可以通过如下条件判断:
- a) $AC-B^2>0$ 存在极值，且当 $A<0$ 有极大值，$A>0$ 时有极小值;
- b) $AC-B^2<0$ 没有极值;
- c) $AC-B^2=0$ 可能有极值，也可能没有极值，需要进一步讨论。

### 有约束条件

```python
import numpy as np
from scipy.optimize import minimize, LinearConstraint, Bounds, NonlinearConstraint
import matplotlib.pyplot as plt
```

```python
# 定义目标函数
def obj_f(x):
    x1 = x[0]
    x2 = x[1]
    obj = -2 * x1 * np.exp(-x1 ** 2 - x2 ** 2)
    return obj
```

```python
x0 = [1, 1]
```

```python
# 定义非线性约束
def nonlinear_c(x):
    x1 = x[0]
    x2 = x[1]
    nlc = np.abs(x1) + np.abs(x2 + 1) - 1
    return nlc
```

```python
nlc = NonlinearConstraint(nonlinear_c, -np.inf, 0) # 定义非线性约束
```

```python
# 定义变量的边界
bounds = Bounds([-1.5, -1.5], [1.5, 1.5])
```

```python
res = minimize(obj_f, x0, method="trust-constr", bounds=bounds, constraints=[nlc])

optimize_x = res.x
```

```python
print("==== Optimal solution ====")
print(res.x)  # 输出优化后的解
```

```python
# 可视化
num = 201
rr = np.linspace(-2, 2, num)
xx1, xx2 = np.meshgrid(rr, rr)
yy = obj_f(np.vstack([xx1.ravel(), xx2.ravel()])).reshape(num, num)

fig, ax = plt.subplots()

ax.contourf(xx1, xx2, yy, levels=20, cmap="RdYlBu_r")

yy_nlc = nonlinear_c(np.vstack([xx1.ravel(), xx2.ravel()])).reshape((num, num))
ax.contour(xx1, xx2, yy_nlc, levels=[0], colors="k")

plt.plot(optimize_x[0], optimize_x[1], "rx", markersize=12)

ax.set_xlabel("$\it{x}_1$")
ax.set_ylabel("$\it{x}_2$")
ax.axis("square")
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
ax.set_xlim(rr.min(), rr.max())
ax.set_ylim(rr.min(), rr.max())
```
