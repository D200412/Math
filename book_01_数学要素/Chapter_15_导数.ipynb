## 15.2 极限：研究微积分的重要数学工具

### 函数极限

设函数 $f(x)$ 在点 $a$ 的某一个去心邻域内有定义，如果存在常数 $C$，对于任意给定正数 $\varepsilon $，不管它多小，总存在正数 $\delta $，使得 $x$ 满足如下不等式时，
$$
0<|x-a|<\delta
$$
对应函数值 $f(x)$ 都满足
$$
|f(x)-C|<\varepsilon
$$
常数 $C$ 就是函数 $f(x)$ 当 $x \to a$ 时的极限，记做:
$$
\lim_{x \to a}f(x)=C
$$

### 邻域

<font color="blue">邻域</font>(neighbourhood)实际上就是一个特殊的开区间。点 $a$ 的 $h(h>0)$ 邻域满足 $a-h<x<a+h$。

$a$ 为邻域的中心，$h$ 为邻域的半径。而<font color="blue">去心邻域</font>(deleted neighborhood 或 punctured neighborhood)指的是，在 $a$ 的邻域中去掉 $a$ 的集合。

```python
from sympy import latex, lambdify, limit, log, oo
from sympy.abc import x
import numpy as np
from matplotlib import pyplot as plt
```

```python
f_x = (1+ 1/x)**x
f_x
```

```python
x_array = np.linspace(0.1, 1000, 1000)
f_x_fcn = lambdify(x, f_x)
f_x_array = f_x_fcn(x_array)
f_x_oo_limit = limit(f_x, x, oo)

fig, ax = plt.subplots()
ax.plot(x_array, f_x_array, linewidth = 1.5)
ax.axhline(y=f_x_oo_limit, color="r")
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5,0.5])
ax.set_xlim(x_array.min(), x_array.max())
plt.xscale("log")
ax.set_xlabel(r"$\it{x}$")
ax.set_ylabel(r"$%s$" % latex(f_x))

plt.grid(True, which="both", ls="-")
```

## 15.3 左极限、右极限

### 右极限 

将极限条件改为 $x \to a^{+}$，$C$ 叫做函数 $f(x)$ 的<font color="blue">右极限</font>(right-hand limit 或 right limit)，记做:

$$
\lim_{x \to a^{+}}f(x)=C
$$

### 左极限

将极限条件改为 $x \to a^{-}$，$C$ 叫做函数 $f(x)$ 的<font color="blue">左极限</font>(left-hand limit 或 left limit)，记做:
$$
\lim_{x \to a^{-}}f(x)=C
$$

### 极限不存在

$\displaystyle\lim_{x \to a}f(x)$不存在可能有三种情况:
- $f(x)$ 在 $x=a$ 处左右极限不一致；
- $f(x)$ 在 $x=a$ 处趋向无穷；
- $f(x)$ 在趋向 $x=a$ 时在两个定值之间震荡。


```python
from sympy import latex, lambdify, limit, log, oo
from sympy.abc import x
import numpy as np
from matplotlib import pyplot as plt
```

```python
f_x = 1 / (1 + 2 **(-1/x))
f_x_fcn = lambdify(x, f_x)
x_array_right = np.linspace(0.01, 4, 500)
f_x_array_right = f_x_fcn(x_array_right)
f_x_0_limit_right = limit(f_x, x, 0, "+") # 1

x_array_left = np.linspace(-4, -0.01, 500)
f_x_array_left = f_x_fcn(x_array_left)
f_x_0_limit_left = limit(f_x, x, 0, "-") # 0

fig, ax = plt.subplots()
ax.plot(x_array_right, f_x_array_right, linewidth=1.5, color="b")
ax.axhline(y=f_x_0_limit_right, color="r")

ax.plot(x_array_left, f_x_array_left, linewidth=1.5, color="b")
ax.axhline(y=f_x_0_limit_left, color="r")

ax.axvline(x=0, color="k")
ax.axhline(y=0.5, color="k")
ax.grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])
ax.set_xlim(x_array_left.min(), x_array_right.max())

ax.set_xlabel(r"$\it{x}$")
ax.set_ylabel(r"$%s$" % latex(f_x))
```

### 函数导数定义

对于函数 $y=f(x)$ 自变量 $x$ 在 $a$ 点处一个微小增量 $\bigtriangleup x$，会导致函数值增量 $\bigtriangleup y=f(a+\bigtriangleup x)-f(a)$。

当 $\bigtriangleup x$ 趋向于 0 时，函数值增量 $\bigtriangleup y$ 和自变量增量 $\bigtriangleup x$ 比值的极限存在，则称 $y=(x)$ 在 $a$ 处可导(functionfof x is differentiable at a)。这个极限值便是函数 $f(x)$ 在 $a$ 点处一阶导数值:
$$
f'(a) = f'(x)|_{x=a} = \frac{\mathrm{d}f(x)}{\mathbf{d}x}|_{x=a} = \lim_{\bigtriangleup x \to 0}\frac{f(a + \bigtriangleup x) - f(a)}{\bigtriangleup x}
$$
从几何角度看，随着 $\bigtriangleup x$ 不断减小，割线不断接近切线。


<font color="green"> $\bigtriangleup$ 和 $\mathrm{d}$ 都是“差”(difference)的含义。但是，$\bigtriangleup$ 代表近似值，比如 $\bigtriangleup x \to 0$；而 $\mathrm{d}$ 是精确值，比如$\mathrm{d}x$。白话说，$\mathrm{d}x$ 是 $\bigtriangleup x$ 趋向于 $0$ 的精确值。</font>

如果函数 $y=f(x)$ 在 $x=a$ <font color="blue">可导</font>(differentiable)则函数在该点处<font color="blue">连续</font>(continuous)；但是，函数在某一点处连续并不意味着函数可导。

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm
from sympy import latex, lambdify
from sympy.abc import x
```

```python
def plot_scant(x0, y0, x1, y1, color):
    # 计算割线的斜率k
    k = (y1 - y0) / (x1 - x0)
    x = np.linspace(-1, 4, 100)
    secant_y_x = k * (x  - x0)  + y0
    plt.plot(x, secant_y_x, color=color, linewidth=0.25)
```

```python
delta_Xs = np.linspace(0.1, 1, 10)
f_x = x**2
x_array = np.linspace(-1, 4, 100)
f_x_fcn = lambdify(x, f_x)
y_array = f_x_fcn(x_array)

x0 = 1
y0 = f_x_fcn(x0)

fig, ax = plt.subplots(figsize=(8,8))

plt.plot(x_array, y_array, color="#00448A", linewidth=1.25)
plt.plot(x0, y0, color="#92D050", markersize=12)

colors = plt.cm.RdYlBu_r(np.linspace(0, 1, len(delta_Xs)))

for i in np.linspace(1, len(delta_Xs), len(delta_Xs)):
    x1 = x0 + delta_Xs[int(i) - 1]
    y1 = f_x_fcn(x1)
    plt.plot(x1, y1, color="#00448A", marker="x", markersize=12)
    plot_scant(x0, y0, x1, y1, colors[int(i) - 1])

plt.xlabel("X")
plt.ylabel("$y = f(x)$")
ax.set_title("$f(x) = %s$" % latex(f_x))
ax.set_xlim(0, 2)
ax.set_ylim(-1, 4)
```

### 驻点

对于一元函数 $f(x)$，<font color="blue">驻点</font>(stationarypoint)是函数一阶导数为 0 的点。

驻点可能是一元函数的极小值、极大值、鞍点。

```python
from sympy import latex, lambdify, diff, sin
from sympy.abc import x
import numpy as np
from matplotlib import pyplot as plt
```

```python
# 原函数
f_x = x ** 2 - 2
x_array = np.linspace(-2, 2, 100)
f_x_fcn = lambdify(x, f_x)
f_x_array = f_x_fcn(x_array)

# 一阶导数
f_x_1_diff = diff(f_x, x)
f_x_1_diff_fcn = lambdify(x, f_x_1_diff)
f_x_1_diff_array = f_x_1_diff_fcn(x_array)

# 二阶导数
f_x_2_diff = diff(f_x, x, 2)
f_x_2_diff_fcn = lambdify(x, f_x_2_diff)
f_x_2_diff_array = f_x_2_diff_fcn(x_array)
f_x_2_diff_array = f_x_2_diff_array + x_array * 0  # 确保数组形状与x_array一致

fig, ax = plt.subplots(3, 1)

# 绘制原函数
ax[0].plot(x_array, f_x_array, linewidth=1.5)
ax[0].hlines(y=0, xmin=x_array.min(), xmax=x_array.max(), color="r", linestyle="--")
ax[0].vlines(x=0, ymin=y_array.min(), ymax=y_array.max(), color="r", linestyle="--")
ax[0].set_title("$f(x) = %s$" % latex(f_x))
ax[0].set_ylabel("$f(x)$")
ax[0].set_xlim(x_array.min(), x_array.max())
ax[0].spines["right"].set_visible(False)
ax[0].spines["top"].set_visible(False)
ax[0].set_xticklabels([])
ax[0].grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])

# 绘制一阶导数
ax[1].plot(x_array, f_x_1_diff_array, linewidth=1.5)
ax[1].hlines(y=0, xmin=x_array.min(), xmax=x_array.max(), color="r", linestyle="--")
ax[1].vlines(x=0, ymin=f_x_1_diff_array.min(), ymax=f_x_1_diff_array.max(), color="r", linestyle="--")
ax[1].set_title(r"$f\'(x) = %s$" % latex(f_x_1_diff))
ax[1].set_xlabel(r"$\it{x}$")
ax[1].set_ylabel(r"$f\'(x)$")
ax[1].set_xlim(x_array.min(), x_array.max())
ax[1].spines["right"].set_visible(False)
ax[1].spines["top"].set_visible(False)
ax[1].grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])

# 绘制二阶导数
ax[2].plot(x_array, f_x_2_diff_array, linewidth=1.5)
ax[2].hlines(y=0, xmin=x_array.min(), xmax=x_array.max(), color="r", linestyle="--")
ax[2].vlines(x=0, ymin=f_x_2_diff_array.min(), ymax=f_x_2_diff_array.max(), color="r", linestyle="--")
ax[2].set_title(r"$f^{(2)}(x) = %s$" % latex(f_x_2_diff))
ax[2].set_xlabel(r"$\it{x}$")
ax[2].set_ylabel(r"$f\'\'(x)$")
ax[2].set_xlim(x_array.min(), x_array.max())
ax[2].spines["right"].set_visible(False)
ax[2].spines["top"].set_visible(False)   
ax[2].grid(linestyle="--", linewidth=0.25, color=[0.5, 0.5, 0.5])

plt.tight_layout()            
```
